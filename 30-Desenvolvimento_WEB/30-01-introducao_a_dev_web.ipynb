## ***O que é?***

O desenvolvimento web é o processo de construir e manter sites e aplicações que rodam no navegador. Se pensarmos em um prédio, o desenvolvimento web envolve desde a fundação e fiação (que ninguém vê) até a pintura e decoração (o que todos usam).

Python entra nessa história como uma ferramenta poderosa para o "motor" dessas aplicações. Vamos decompor isso:

----------------------------------------------------------------------------------------------------

### ***Tipos de Componentes Principais***

Existem três áreas principais: front-end (interface visual do usuário), back-end (lógica do servidor e bancos de dados) e full-stack (combinação das duas).

- Front-end foca no que o usuário vê e interage, como layouts responsivos (Ex.: HTML, CSS e JavaScript).

- Back-end gerencia dados, autenticação e servidores (Ex.: Python e Java).

- Full-stack permite atuar em todo o projeto, sendo muito valorizado.

A situação representada acima, é feita para nos ajudar a entender a importância das coisas.

Obs.: Porém, não quer dizer que sempre será nessa proporção. Há situações em que há mais de Frontend para ser feito.

### ***Os Principais Frameworks***

Frameworks são coleções de ferramentas que evitam que você precise "reinventar a roda". Em Python, os dois titãs são:

- Django (A abordagem "Bateria Inclusa")
O Django foca em rapidez e segurança. Ele já vem com quase tudo o que você precisa: sistema de login, painel administrativo e conexão com banco de dados. É ideal para projetos grandes e complexos.

- Flask (A abordagem "Minimalista")
O Flask é leve e flexível. Ele te dá o básico e deixa você escolher quais outras ferramentas quer adicionar. É perfeito para iniciantes ou para aplicações menores (microserviços).

- FastAPI (O "Veloz")
O queridinho do momento. É focado em alta performance e na criação de APIs (pontes de comunicação entre sistemas), sendo extremamente rápido e moderno.

## ***O Fluxo de Trabalho***

Quando você usa Python para criar um site, o processo geralmente segue este caminho:

1. Requisição: O usuário clica em um botão no navegador (Frontend).

2. Roteamento: O servidor Python recebe o pedido e decide qual função deve ser executada.

3. Lógica/Banco de Dados: O Python busca informações no banco de dados (ex: "Qual o saldo desse usuário?").

4. Resposta: O Python envia os dados de volta, geralmente formatados para que o Frontend os exiba.

## ***Internet vs. Web***

## 1. A Internet (A Infraestrutura)
A Internet é a rede física de redes. Ela é a fundação, o hardware e os protocolos que permitem que bilhões de dispositivos se conectem.

O que compõe: Cabos submarinos de fibra ótica, satélites, roteadores, Wi-Fi e os próprios computadores/celulares.

Protocolo Base: O TCP/IP. Ele é o "idioma" básico que permite que um pacote de dados saia do Brasil e chegue ao Japão.

O que viaja por ela: Tudo! Não só sites, mas também e-mails, chamadas de WhatsApp, jogos online, streaming de câmeras de segurança e transferências de arquivos (FTP).

Analogia: A Internet são as ruas, avenidas e trilhos espalhados pelo mundo todo.

## 2. A Web ou WWW (O Serviço)
A World Wide Web (Web) é apenas um dos muitos serviços que rodam em cima da Internet. Ela é uma imensa coleção de documentos e recursos conectados por links (hiperlinks).

O que compõe: Páginas HTML, arquivos CSS, imagens e scripts.

Protocolo Base: O HTTP (ou HTTPS). É a regra específica para pedir e receber páginas web.

Como acessamos: Através de um Navegador (Chrome, Firefox, Safari). O navegador é o "tradutor" da Web.

Analogia: A Web são os caminhões e carros de entrega que usam as estradas para levar informações até você.

## ***Por que isso importa para o Desenvolvedor?***

Quando você está programando em Python com Flask ou Django, você está trabalhando na Web. Você está criando regras (em HTTP) para que, quando alguém use a Internet para chegar até o seu servidor, o seu código saiba qual "pacote de informação" devolver para o navegador do usuário.
