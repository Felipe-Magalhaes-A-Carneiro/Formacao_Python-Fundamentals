## ***1. O que é o FastAPI?***
O FastAPI é um framework Python moderno, de altíssima performance, focado exclusivamente na criação de APIs. Ele foi criado para aproveitar os recursos mais recentes do Python (como a programação assíncrona e a tipagem de dados).

### ***Os 3 Pilares do FastAPI:***
1. Velocidade: É um dos frameworks Python mais rápidos que existem (chega perto do Node.js e Go).

2. Tipagem (Pydantic): Ele usa os "types" do Python para validar os dados. Se você espera um número e o usuário envia um texto, o FastAPI barra o erro antes mesmo de chegar na sua lógica.

3. Documentação Automática: Ele cria uma página interativa para testar sua API sozinho.

----------------------------------------------------------------------------------------------------

## ***2. Exemplo Prático: Sua primeira API "Turbo"***

Imagine que queremos uma API para calcular o dobro de um número. Veja como o código é elegante:

```python
from fastapi import FastAPI
from pydantic import BaseModel

app = FastAPI()

# Documentação automática disponível em /docs

@app.get("/")
def raiz():
    return {"mensagem": "Bem-vindo à minha API veloz!"}

@app.get("/dobro/{numero}")
def calcular_dobro(numero: int):
    # O ': int' garante que o FastAPI só aceite números
    resultado = numero * 2
    return {"original": numero, "resultado": resultado}
```

### ***O "Pulo do Gato": Documentação Interativa***

Ao rodar esse código, o FastAPI gera automaticamente um endereço chamado `/docs`. Lá, você encontra o Swagger UI: uma página onde você pode clicar nos botões e testar sua própria API sem precisar de ferramentas externas.

## ***3. Por que ele é "Moderno"? (Async/Await)***

O FastAPI brilha com o Asynchronous Code.

- Frameworks antigos (como o Flask padrão): Se 10 pessoas pedem algo ao mesmo tempo, o servidor atende uma por uma (ou cria várias cópias de si mesmo, consumindo muita memória).

- FastAPI: Ele consegue "pausar" um pedido que está esperando o Banco de Dados e atender outro enquanto isso. É como um garçom que anota o pedido de uma mesa e, enquanto o prato não sai, já vai servindo as bebidas de outra. Logo, é ótima para lidar com "IO" (ENTRADA e SAÍDA INPUT e OUTPUT)

## ***3.1 Pontos negativos?***

O FastAPI brilha com o Asynchronous Code.

- Comunidade menor: Embora esteja crescendo, sua comudade ainda é menor em comparação com frameworks estavelecidos como Django.

- Maturidade: Por ser relativamente novo, pode faltar alguma maturidade e ter menos plugins quando comparado a outros frameworks mais antigos.

- Complexidade em projetos grandes: Em projetos muito grandes e complexos, a gestão pode se tornar um desafio, especialmente para quem não esta acostumado com a tpagem estrita e a programação assíncrona.

----------------------------------------------------------------------------------------------------

## ***4. Quando escolher o FastAPI?***

- Quando você quer construir o backend de um aplicativo mobile ou de um site em React/Vue.

- Quando performance é a prioridade número 1.

- Para projetos baseados em Microserviços (pequenas partes de um sistema que se comunicam).
