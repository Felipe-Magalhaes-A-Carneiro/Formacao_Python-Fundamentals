----------------------------------------------------------------------------------------------------

é, em termos simples, um garçom.

Imagine que você está em um restaurante:

- ***Você (O Cliente):*** É o computador ou celular que quer algo (ex: ver a previsão do tempo).

- ***A Cozinha (O Sistema):*** É onde os dados estão guardados e a lógica acontece.

- ***O Garçom (A API):*** É quem leva o seu pedido até a cozinha e traz o prato (os dados) de volta para você. Você não precisa saber como o chef cozinha; você só precisa saber como pedir ao garçom.

----------------------------------------------------------------------------------------------------

## ***1. Como uma API funciona na prática?***

Uma API define um conjunto de regras que permite que dois softwares "conversem". Ela funciona através de "pontos de acesso" chamados Endpoints.

Se o site do Instagram fosse uma casa, os endpoints seriam:

- `/postar_foto`: Um ponto de acesso para enviar imagens.

- `/ver_perfil`: Um ponto de acesso para ler dados de um usuário.

----------------------------------------------------------------------------------------------------

## ***2. Os 4 Verbos Principais (Métodos HTTP)***

Toda interação com uma API usa um "verbo" que diz o que você quer fazer. No mundo técnico, chamamos isso de CRUD:

1. ***`GET` (Read):*** "Me dê essa informação." (Ex: Ver o saldo no app do banco).

2. ***`POST` (Create):*** "Crie um novo registro." (Ex: Fazer um post novo no Twitter).

3. ***`PUT/PATCH` (Update):*** "Altere algo que já existe." (Ex: Mudar sua foto de perfil).

4. ***`DELETE` (Delete):*** "Apague isso." (Ex: Excluir um e-mail).

----------------------------------------------------------------------------------------------------

## ***3. O "Idioma" da API: O JSON***

Antigamente, as APIs usavam formatos complexos (como XML). Hoje, quase todas falam JSON. É um formato de texto leve e fácil de ler.

Exemplo de uma resposta de API de clima:

```python
{
  "cidade": "São Paulo",
  "temperatura": "25°C",
  "condicao": "Ensolarado"
}
```

O seu código Python lê esse texto e decide: "Vou desenhar um sol na tela do usuário".

## ***4. Por que as APIs são o coração da internet moderna?***

Sem APIs, a internet como conhecemos deixaria de existir. Elas permitem a integração:

- ***Login com Google/Facebook:*** O site que você está usando "pergunta" para a API do Google se você é você mesmo.

- ***Pagamentos com Pix/Cartão:*** A loja virtual usa a API do banco ou do Gateway de pagamento para processar a transação.

- ***Google Maps em outros apps:*** O iFood não criou um mapa do zero; ele usa a API do Google Maps dentro do app dele.

## ***5. Tipos de APIs***

- ***APIs Públicas:*** Qualquer desenvolvedor pode usar (ex: API do PokeAPI para ver dados de Pokémon).

- ***APIs Privadas:*** Usadas apenas dentro de uma empresa (ex: a API que o RH usa para falar com o sistema de Folha de Pagamento).

- ***APIs de Parceiros:*** Apenas empresas parceiras têm a "chave" para acessar.

## ***7. API RESTful***

RESTfuls refere-se a APIs que seguem os princípios do REST (Representational State Transfer). São baseadas em padrões HTTP e utilizadas para interações web.

### Características de APIs RESTful

- Uso dos métodos HTTP (GET, POST, PUT, DELETE) para operação CRUD (CREATE, READ, UPDATE, DELETE).
- Curva de aprenzado menor.
- Fácil de entender e implementar.

----------------------------------------------------------------------------------------------------

## ***8. API SOAP***

SOAP (Simple Object Access Protocol) é um protocolo que define um padrão para a troca de mensagens baseadas em XML.

https://www.w3schools.com/xml/xml_soap.asp

### Características de APIs RESTful

- Protocolo baseado em XML para troca de informações.
- Independente de linguagem e plataforma de transporte.
- Suporte para operações complexas e segurança avançado.

----------------------------------------------------------------------------------------------------

## ***9. API GraphQL***

Uma linguagem de consulta para sua API, e um servidor capaz de executar essas consultas, retornando apenas os dados especificados.

### Características de APIs GraphQL

- Permite que os clientes especifiquem exatamente quais dados querem.
- Eficiente na redução de solicitações e no tamanho dos dados transferidos.
- Flexível e fortemente tipada, facilitando a evolução das APIs.

----------------------------------------------------------------------------------------------------

## ***10. Escolhendo o tipo certo de API***

A escolha depende das necessidades especificas do projeto, dos recursos disponíveis e da expertise da equipe.

- RESTful é popular pela simplicidade;
- SOAP é preferido para segurança e transações complexas;
- GraphQL é ideal para aplicações que requerem dados dinâmicos personalizados.
