# ***Atributos***

----------------------------------------------------------------------------------------------------

Atributos são as características ou propriedades de um objeto. Se o objeto fosse um substantivo (ex: Carro), os atributos seriam os adjetivos (ex: cor, modelo, ano).

No Python, existem dois tipos de atributos que você precisa conhecer para não se confundir:

----------------------------------------------------------------------------------------------------

## ***1. Atributos de Instância***

Estes são os mais comuns. Eles pertencem a cada objeto específico. Se você criar dois carros, cada um pode ter uma cor diferente. Eles são definidos quase sempre dentro do método __init__ usando o prefixo self.

```python
class Carro:
    def __init__(self, marca, cor):
        self.marca = marca  # Atributo de instância
        self.cor = cor      # Atributo de instância

c1 = Carro("Ferrari", "Vermelho")
c2 = Carro("Fusca", "Azul")

print(c1.cor) # Saída: Vermelho
print(c2.cor) # Saída: Azul (cada um tem o seu)
```

Ou seja, é como fizemos nas aulas anteriores. Nada novo aqui.

----------------------------------------------------------------------------------------------------

## ***2. Atributos de Classe***

Estes pertencem à Classe como um todo, e não a um objeto específico. 

Todas as instâncias compartilham o mesmo valor. Se você mudar o atributo de classe, ele muda para todo mundo. Eles são definidos diretamente no corpo da classe, fora de qualquer método.

```python
class Carro:
    rodas = 4  # Atributo de classe (todo carro nasce com 4 rodas)

    def __init__(self, marca):
        self.marca = marca

carro_1 = Carro("Ferrari")
carro_2 = Carro("Fusca")

print(carro_1.rodas) # 4
print(carro_2.rodas) # 4
```

----------------------------------------------------------------------------------------------------

## ***3. Atributos Dinâmicos (A "Mágica" do Python)***

Diferente de linguagens rígidas como Java ou C#, o Python permite que você crie atributos "no ar", mesmo depois que o objeto já foi criado.

```python
carro_1 = Carro("Ferrari")
carro_1.teto_solar = True # Criei um atributo que só o carro_1 tem!

print(carro_1.teto_solar) # True
# print(carro_2.teto_solar) # Erro! carro_2 não tem esse atributo.
```

----------------------------------------------------------------------------------------------------

## ***4. Onde os atributos ficam guardados?***

Por baixo dos panos, o Python guarda os atributos de instância em um dicionário chamado __dict__. Você pode até espiar o que tem dentro do seu objeto:

```python
print(carro_1.__dict__) 
# Saída: {'marca': 'Ferrari', 'teto_solar': True}
```

----------------------------------------------------------------------------------------------------

## ***Resumo: Atributo vs. Variável***

- ***Variável comum:*** Existe apenas dentro de uma função ou script (some quando a função acaba).

- ***Atributo:*** Está "colado" ao objeto. Enquanto o objeto existir na memória, o atributo estará lá com ele.

----------------------------------------------------------------------------------------------------

## ***Uma dúvida comum: "Método é um atributo?"***

Tecnicamente, em Python, sim! Métodos são atributos funcionais. A diferença é que os atributos guardam dados (estados) e os métodos guardam comportamentos (ações).

# ***Como proteger esses atributos para que ninguém mude por fora da classe?***

----------------------------------------------------------------------------------------------------

Para proteger os atributos e garantir que ninguém mude qualquer outro dado sensível de forma desordenada, utilizamos o ***Encapsulamento*** combinado com o ***decorador @property***.

No Python, seguimos a convenção de "proteger" o atributo e oferecer apenas o caminho oficial para acessá-lo.

Obs.: Falamos primeiramente sobre @property em:

[09-07-POO-encapsulamento_property](<../09-POO com Python/09-07-POO-encapsulamento_property.ipynb>)

Se quiser saber mais sobre ele.

----------------------------------------------------------------------------------------------------

## ***1. Tornando o Atributo "Privado"***

O primeiro passo é mudar o nome do atributo no _ _init_ _ para usar dois underlines (__). Isso aciona o ***'Name Mangling'***, dificultando o acesso direto de fora.

```python
class Carro:
    def __init__(self, marca):
        self.__marca = marca  # Agora ele está "escondido"
```

----------------------------------------------------------------------------------------------------

## ***2. Criando o Acesso de "Apenas Leitura"***

Se você quer que as pessoas vejam a marca, mas não consigam alterá-la, você cria apenas o ***Getter (o @property)***. 

Como não vamos criar o Setter, a marca se torna imutável por fora.

```python
class Carro:
    def __init__(self, marca):
        self.__marca = marca

    @property
    def marca(self):
        return self.__marca

# Testando:
meu_carro = Carro("Ferrari")
print(meu_carro.marca)  # Funciona: retorna "Ferrari"
```

A classe esta protegida.

Vamos provar isso tentando altera-la:

```python
# Tentando mudar:
meu_carro.marca = "Fiat"  # Erro! AttributeError: can't set attribute
```

O erro mostra que o método esta, de fato, protegito.

----------------------------------------------------------------------------------------------------

## ***3. Protegendo com Validação (O Setter)"***

E se você permitir a mudança, mas quiser garantir que ela seja válida? 

Usamos o .setter. Imagine que o carro não pode ter uma marca vazia:

```python
class Carro:
    def __init__(self, marca):
        self._marca = marca # Usando um underline (protegido por convenção)

    @property
    def marca(self):
        return self._marca

    @marca.setter
    def marca(self, nova_marca):
        if not nova_marca.strip():
            print("Erro: A marca não pode ser um texto vazio!")
        else:
            print(f"Marca alterada de {self._marca} para {nova_marca}")
            self._marca = nova_marca

c = Carro("Ferrari")
c.marca = " "       # Saída: Erro: A marca não pode ser um texto vazio!
c.marca = "Porsche" # Funciona!
```

----------------------------------------------------------------------------------------------------

## ***Por que isso é melhor que o acesso direto?***

1. Controle de Qualidade: Você impede que dados "sujos" ou inválidos entrem no seu objeto.

2. Segurança de Negócio: Algumas coisas (como o CPF de um cliente ou o ID de um produto) nunca devem mudar depois de criadas.

3. Rastreabilidade: Você pode colocar um print ou um log dentro do @property para saber toda vez que alguém tentar ler ou mudar aquele valor.

----------------------------------------------------------------------------------------------------

## ***Resumo do Kit de Proteção:***

***self.nome:*** Público (Porta aberta).

***self._nome:*** Protegido (Porta encostada - "Por favor, não entre").

***self.__nome:*** Privado (Porta trancada com Name Mangling).

***@property:*** O visor da porta (Permite ver sem entrar).

***@nome.setter:*** O segurança na porta (Só entra quem segue as regras).
