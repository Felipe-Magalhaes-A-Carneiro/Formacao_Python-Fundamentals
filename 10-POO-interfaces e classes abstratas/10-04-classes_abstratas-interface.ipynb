Neste assunto, precisamos falar sobre os "contratos" do Python. 

Se a ***Herança*** diz que um filho recebe algo do pai, as 'Classes Abstratas' e 'Interfaces' dizem que ***o filho é obrigado*** a fazer algo.

No Python, ambos os conceitos são realizados através do módulo abc (Abstract Base Classes). Ou seja, uma biblioteca externa do Python.

----------------------------------------------------------------------------------------------------

## ***1. O que é uma Classe Abstrata?***

***Uma classe abstrata é uma classe que não pode ser instanciada*** (você não pode criar um objeto dela). 

***Ela serve apenas como um modelo ou "rascunho" para outras classes.***

Imagine a classe FormaGeometrica. Não faz sentido criar um objeto "Forma", mas faz sentido criar um "Quadrado" ou "Círculo" que herdam dela.

- Método Abstrato: É um método que a classe pai declara, mas não implementa. Ela diz: 
***"Quem me herdar, tem que escrever sua própria versão desse método".***

```python
from abc import ABC, abstractmethod

class Forma(ABC): # Herdar de ABC torna a classe abstrata
    @abstractmethod
    def calcular_area(self):
        pass
```

Como foi dito anteriormente, 'classes abstratas' não permitem que um objeto seja criado por meio dela.

Vamos tentar:

```python
circulo = Forma() # Isso daria erro!
```

***TypeError: Can't instantiate abstract class Forma without an implementation for abstract method 'calcular_area'*** O erro esta correto!

----------------------------------------------------------------------------------------------------

## ***2. O que é uma Interface?***

Em linguagens como Java, a 'Interface' é uma estrutura separada. No Python, não existe uma palavra-chave interface. 

***Nós usamos classes abstratas que contêm apenas métodos abstratos para simular uma interface.***

A interface define um comportamento. Por exemplo: tudo que é Autenticavel precisa ter o método fazer_login(), seja um Usuario, um Admin ou um Robo.

----------------------------------------------------------------------------------------------------

## ***3. Diferença na Prática***

| Característica | Característica | Característica |
| :--- | :---: | ---: |
| Objetivo | Compartilhar código e definir estrutura. | Definir um contrato de comportamento. |
| Métodos | Pode ter métodos reais (com código) e abstratos. | Geralmente só tem métodos abstratos (pass). |
| Atributos | Pode ter atributos de instância e de classe. | Geralmente não possui atributos. |
| Relação | "O que o objeto é" (Animal, Veículo). | "O que o objeto consegue fazer" (Voável, Carregável).0 |


----------------------------------------------------------------------------------------------------

## ***4. Por que usar isso? (O "Contrato")***

O maior benefício é a segurança. Se você tentar criar uma subclasse de uma Classe Abstrata e esquecer de implementar um método "@abstractmethod", como nós fizemos no primeiro código acima, o Python vai impedir o programa de rodar até que você conserte.

```python
class Quadrado(Forma):
    def __init__(self, lado):
        self.lado = lado
    
    # Se eu esquecer de definir calcular_area(), o Python trava.
    def calcular_area(self):
        return self.lado ** 2
```
