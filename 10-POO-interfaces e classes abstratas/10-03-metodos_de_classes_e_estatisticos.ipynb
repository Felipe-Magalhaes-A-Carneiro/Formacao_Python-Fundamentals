# ***Métodos de Classe e Métodos Estáticos***

----------------------------------------------------------------------------------------------------

Além dos métodos comuns (de instância), o Python oferece duas formas de criar métodos que não dependem de um objeto específico, mas sim da Classe.

A grande diferença entre eles é o que eles conhecem sobre a classe.

----------------------------------------------------------------------------------------------------

## ***1. Métodos de Classe (@classmethod)***

Um método de classe recebe a própria classe como primeiro parâmetro (por convenção, chamado de cls). Ele consegue acessar e modificar os Atributos de Classe que vimos antes.

- Para que serve? Geralmente usado para criar "construtores alternativos" (fábricas de objetos).

- cIdentificador: Usa o decorador ***@classmethod***.

```python
class Robo:
    populacao = 0  # Atributo de classe

    def __init__(self, nome):
        self.nome = nome
        Robo.populacao += 1

    @classmethod
    def ler_populacao(cls):
        # cls refere-se à classe Robo
        print(f"Total de robôs fabricados: {cls.populacao}")

# Chamamos direto pela Classe, sem precisar de um objeto
Robo.ler_populacao()
```

----------------------------------------------------------------------------------------------------

## ***2. Métodos Estáticos (@staticmethod)***

Um método estático é como uma função comum que "mora" dentro da classe por uma questão de organização. Ele não recebe nem self nem cls. Ele não sabe nada sobre os atributos da classe ou do objeto.

- Para que serve? Para funções utilitárias que fazem sentido estar ali dentro, mas que não precisam de dados da classe para trabalhar.

- Identificador: Usa o decorador ***@staticmethod***.

```python
class CalculadoraEspacial:
    @staticmethod
    def converter_anos_luz(km):
        # Apenas um cálculo matemático puro
        return km / 9.461e12

# Uso:
distancia = CalculadoraEspacial.converter_anos_luz(1000000)
```
