## ***1. O que é um gerador?***

Um gerador é uma função especial que retorna um iterator lazy (lazy iterator) — um objeto que produz valores sob demanda, um de cada vez, sem armazenar toda a sequência na memória.

Geradores são como listas mágicas que criam itens conforme necessário, usando yield.

### ***1.1. Diferença fundamental:***

| Função Normal                            | Gerador                                     |
| ---------------------------------------- | ------------------------------------------- |
| return - retorna valor e termina | yield - retorna valor e pausa            |
| Executa toda a lógica                    | Executa sob demanda  |
| Armazena tudo na memória                 | Gera um por vez                          |
| Pode ser chamada várias vezes            | Single-use (consumido uma vez)       |

## ***2. O poder do yield***

### ***2.1. Como funciona yield***

```python
def contador(n):
    print("Função começando")
    i = 0
    while i < n:
        print(f"Yielding {i}")
        yield i  # pausa aqui e retorna valor
        i += 1
    print("Função terminando")

# Usando
for numero in contador(3):
    print(f"Recebido: {numero}")
```

O que acontece:

- Cada yield pausa a função

- Mantém o estado atual (variáveis, posição)

- Retorna o valor

- Na próxima chamada, continua de onde pausou

### ***2.2. Comparação: return vs yield***

```python
# Função normal - cria lista inteira
def numeros_com_return(max_number):
    lista = []
    for number in range(max_number + 1):
        lista.append(number)
    return lista  # retorna tudo junto

result = numeros_com_return(1000000)  # lista de 1M elementos na memória!
print(len(result))  



# Gerador - gera um por vez
def numeros_com_yield(max_number):
    for number in range(max_number + 1):
        yield number  # entrega um e pausa

gerador = numeros_com_yield(1000000)  # memória quase zero!
print(next(gerador))  
print(next(gerador)) 
```

## ***3. Criando geradores***

### ***3.1. Função geradora básica***

```python
def fibonacci(limit):
    """Gerador de sequência de Fibonacci"""
    a, b = 0, 1
    for _ in range(limit):
        yield a  # pausa e retorna 
        a, b = b, a + b


for num in fibonacci(10):
    print(num) 
```

### ***3.2. Gerador de números pares***

```python
def pares(até):
    """Gerador de números pares até X"""
    for i in range(0, até + 1, 2):
        yield i

for num in pares(20):
    print(num) 
```

### ***3.3. Gerador que lê arquivo (linhas sob demanda)***

```python
def ler_linhas(arquivo):
    """Gerador que lê arquivo linha por linha"""
    with open(arquivo) as f:
        for linha in f:
            yield linha.strip()  # pausa em cada linha

# Usando com arquivo GRANDE
for linha in ler_linhas('grande_arquivo.txt'):
    print(linha)  
```

## ***4. Expressões Geradoras (Generator Comprehension)***

### ***4.1. Sintaxe simples (como list comprehension)***

```python
# Generator comprehension usa PARENTESES ( ) 
gerador = (x ** 2 for x in range(5))

for num in gerador:
    print(num) 
```

### ***4.2. Comparação: List vs Generator Comprehension***

```python
# List comprehension - cria lista inteira
lista = [x ** 2 for x in range(1000000)]  # 1M elementos na memória!
print(len(lista))  

# Generator comprehension - um por vez
gerador = (x ** 2 for x in range(1000000))  # memória quase zero!
print(next(gerador))  
print(next(gerador))  
```

***Diferença crucial:***

- List comprehension: produz lista inteira de uma vez

- Generator comprehension: produz um item de cada vez → muito mais eficiente!

### ***4.3. Com filtro***

```python
# Gerador com filtro
gerador = (x for x in range(20) if x % 3 == 0)

for num in gerador:
    print(num) 
```
