## ***1. O que é um Interadores?***

Um iterador é um objeto que permite percorrer uma sequência de valores (como lista, tupla, string) um elemento de cada vez, sem precisar saber o tamanho da sequência de antemão.

Iteradores são como um "controle remoto" para navegar em uma coleção de itens.

### ***1.1. Características principais:***

| Característica | Descrição                                                          |
| -------------- | ------------------------------------------------------------------ |
| Unidirecional  | Só avança, não volta ao início sem reiniciar                   |
| Lazy           | Gera valores sob demanda, economizando memória                 |
| Estadual       | Mantém o estado (posição atual) da iteração                   |
| StopIteration  | Gerar exceção quando não há mais elementos  |

## ***3.  Criando um iterador personalizado***

### ***3.1. Exemplo -> Contador de 0 a 4***

```python
class Contador:
    """Iterador que gera números de 0 a 4"""
    
    def __init__(self):
        self.numero = 0
    
    def __iter__(self):
        print("__iter__ chamado")
        return self  # retornará o objeto iterador
    
    def __next__(self):
        print("__next__ chamado")
        if self.numero > 4:
            raise StopIteration  # termina iteração
        resultado = self.numero
        self.numero += 1
        return resultado

# Usando
contador = Contador()

for i in contador:
    print(i)
```

### ***3.2. Exemplo -> Iterador de lista invertida***

```python
class ListaInversa:
    """Iterador que percorre lista do último ao primeiro"""
    
    def __init__(self, dados):
        self.dados = dados
        self.index = len(dados) - 1
    
    def __iter__(self):
        return self
    
    def __next__(self):
        if self.index < 0:
            raise StopIteration
        resultado = self.dados[self.index]
        self.index -= 1
        return resultado

lista = [10, 20, 30, 40]
para_inversa = ListaInversa(lista)

for item in para_inversa:
    print(item)
```

## ***4. Usando iteradores na prática***

### ***4.1. Função iter() - Obtém iterador de um iterável***

```python
lista = [1, 2, 3]

# Obtém iterador da lista
iterador = iter(lista)  # chama __iter__()

print(iterador)  # <list_iterator object>

# Obtém elementos com next()
print(next(iterador))  
print(next(iterador))  
print(next(iterador))  
print(next(iterador))  # StopIteration!
```

### ***4.2. Função next() - Obtém próximo elemento***

```python
lista = ['a', 'b', 'c']
iterador = iter(lista)

# Método 1: next() direto
print(next(iterador))  # 'a'
print(next(iterador))  # 'b'

# Método 2: com valor default (evita StopIteration)
print(next(iterador))  # 'c'
print(next(iterador, 'default'))  # 'default' (não gera erro!)
```

### ***4.3. For loop usa iteradores internamente***

```python
# O que o for faz internamente:
lista = [1, 2, 3]

iterador = iter(lista)  # obtém iterador [web:21]
while True:
    try:
        elemento = next(iterador)  # pega próximo
        print(elemento)
    except StopIteration:  # quando acaba [web:21]
        break
```

## ***5. Iterável vs Iterador (saber a diferença é importante)***

| Conceito | Definição                                              | Exemplo                                                     |
| -------- | ------------------------------------------------------ | ----------------------------------------------------------- |
| Iterável | Objeto que implementa _ _iter_ _() e retorna um iterador | lista, tupla, string, dict pythoniluminado.netlify          |
| Iterador | Objeto que implementa _ _iter_ _() e _ _next_ _()          | iter(lista), objetos com _ _next_ _() pythoniluminado.netlify |

```python
lista = [1, 2, 3]  # É iterável (não é iterador!)

print(hasattr(lista, '__iter__'))  
print(hasattr(lista, '__next__'))  

iterador = iter(lista)  # Agora é iterador

print(hasattr(iterador, '__iter__'))  
print(hasattr(iterador, '__next__')) 
```

Verificando com collections.abc:

```python
from collections.abc import Iterable, Iterator

lista = [1, 2, 3]
iterador = iter(lista)

print(isinstance(lista, Iterable))  # True - é iterável
print(isinstance(lista, Iterator))  # False - não é iterador

print(isinstance(iterador, Iterable))  # True
print(isinstance(iterador, Iterator))  # True - é iterador!
```

## ***5. Geradores (Iteradores automáticos)***

Geradores são iteradores criados automaticamente usando yield.

### ***5.1. Função geradora simples***

```python
def contador(n):
    """Gerador que conta de 0 até n-1"""
    print("Função começando")
    i = 0
    while i < n:
        print(f"Yielding {i}")
        yield i  # pausa e retorna valor
        i += 1
    print("Função terminando")


for numero in contador(3):
    print(f"Recebido: {numero}")
```

### ***5.2. Gerador de Fibonacci***

```python
def fibonacci(limit):
    """Gerador de sequência de Fibonacci"""
    a, b = 0, 1
    for _ in range(limit):
        yield a  # pausa aqui
        a, b = b, a + b

for num in fibonacci(10):
    print(num)  
```

### ***5.3. Expressão geradora (como list comprehension)***

```python
# Gerador expression
gerador = (x ** 2 for x in range(5))

for num in gerador:
    print(num) 
```

## ***6. Geradores vs Iteradores (classe)***

| Característica | Gerador                     | Iterador (classe)                                      |
| -------------- | --------------------------- | ------------------------------------------------------ |
| Criação        | Usando yield em função dio  | Classe com _ _iter_ _ e _ _next_ _ |
| Simplicidade   | Mais simples, menos código  | Mais complexo, mais controle                           |
| Estado         | Mantido automaticamente dio | Deve ser mantido manualmente                           |
| Reutilização   | Geralmentesingle-use        | Pode reiniciar _ _iter_ _                                |

### ***6.1. Comparação na prática***

```python
# Gerador (simples)
def gerador_numeros():
    for i in range(5):
        yield i

# Iterador com classe (mais controle)
class Numeros:
    def __init__(self):
        self.i = 0
    
    def __iter__(self):
        self.i = 0  # pode reiniciar!
        return self
    
    def __next__(self):
        if self.i >= 5:
            raise StopIteration
        self.i += 1
        return self.i - 1
```

## ***7. Iteradores práticos úteis***

### ***7.1. enumerate() - Iterador com índice***

```python
lista = ['a', 'b', 'c']

for indice, valor in enumerate(lista):
    print(f"{indice}: {valor}")
```

### ***7.2. zip() - Iterar múltiplas listas***

```python
nomes = ['Ana', 'Bob', 'Carlos']
idades = [25, 30, 35]

for nome, idade in zip(nomes, idades):
    print(f"{nome} tem {idade} anos")
```

### ***7.3. map() e filter() - Iteradores transformadores***

```python
# map() - transforma cada elemento
numeros = [1, 2, 3, 4]
dobro = map(lambda x: x * 2, numeros)  # é iterador!

print(list(dobro))  # [2, 4, 6, 8]

# filter() - filtra elementos
numeros = [1, 2, 3, 4, 5, 6]
par = filter(lambda x: x % 2 == 0, numeros)  # é iterador!

print(list(par))
```

### ***7.4. Iterar arquivo (printfloat)***

```python
# Arquivo grande não carrega tudo na memória! Obs.: O arquivo não existe

with open('grande_arquivo.txt') as arquivo:
    for linha in arquivo:  # arquivo é iterador!
        print(linha.strip()) 
```

## ***8. Propriedades e limitações***

### ***8.1. Unidirecional (não pode voltar)***

```python
lista = [1, 2, 3, 4, 5]
iterador = iter(lista)

print(next(iterador))  # 1
print(next(iterador))  # 2
# Não pode voltar ao 1!

# Para reiniciar, precisa criar novo iterador
iterador = iter(lista)  
print(next(iterador))  
```

### ***8.2. Não pode verificar se está vazio***

```python
iterador = iter([])

# Não pode saber se está vazio sem consumir!
try:
    primeiro = next(iterador)
    print(f"Tem elemento: {primeiro}")
except StopIteration:
    print("Iterador vazio!")
```

### ***8.3. Single-use (geralmente)***

```python
gerador = (x ** 2 for x in range(3))

print(list(gerador))
print(list(gerador))  
```

## ***10. Benefícios dos iteradores***

- Economia de memória - não carrega toda coleção

- Processamento de dados grandes - arquivo gigante, stream

- Código limpo - loops personalizados sem índices

- Lazy evaluation - calcula sob demanda

- Infinite sequences - sequências infinitas possíveis

```python
# Sequência infinita (só com iterador!)
def infinito():
    i = 0
    while True:
        yield i
        i += 1

cont = infinito()
print(next(cont))  # 0
print(next(cont))  # 1
# ... continua infinitamente
```
