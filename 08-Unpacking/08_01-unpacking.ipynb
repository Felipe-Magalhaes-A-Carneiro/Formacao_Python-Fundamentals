# ***unpacking - "desempacotamento"***

O "desempacotamento" (ou unpacking) é o processo inverso do que vimos antes. Em vez de a função juntar vários itens em uma caixa, você já tem uma caixa (lista ou dicionário) e quer espalhar o conteúdo dela nos parâmetros da função.

## 1. Desempacotando Listas e Tuplas com *

Imagine que você tem uma lista com três números e uma função que recebe três parâmetros. Em vez de passar *lista[0], lista[1], lista[2]*, você usa o asterisco.

```python
def calcular_volume(comprimento, largura, altura):
    return comprimento * largura * altura

dimensoes = [10, 5, 2]

# O asterisco "explode" a lista e passa cada item para um parâmetro
resultado = calcular_volume(*dimensoes)

print(resultado) # Saída: 100
```

## 2. Desempacotando Dicionários com **
Isso é muito comum em sistemas de configurações. Se você tem um dicionário onde as chaves têm exatamente os mesmos nomes dos parâmetros da função, o ** faz a mágica acontecer.

```python
def criar_usuario(nome, idade, cargo):
    print(f"Salvando: {nome}, {cargo}, {idade} anos.")

dados_da_web = {
    "nome": "Carla",
    "idade": 30,
    "cargo": "Engenheira"
}

# Os dois asteriscos dizem: "Use as chaves do dict como nomes de parâmetros"
criar_usuario(**dados_da_web)
```

## 3. Por que isso é útil?

- Limpeza: O código fica muito mais curto e fácil de ler.

- Integração: Se você recebe dados de um banco de dados ou de uma API (que geralmente vêm como dicionários), você pode passá-los direto para suas funções sem precisar mapear um por um.

```python
import random

def sortear_premiado(premio, *nomes):
    try:
        if len(nomes) < 2:
            raise ValueError("É necessário pelo menos 2 nomes para o sorteio!")
            
        vencedor = random.choice(nomes)
        print(f"O prêmio '{premio}' vai para: {vencedor}!")
        
    except ValueError as erro:
        print(f"Erro no sorteio: {erro}")

# Temos uma lista de candidatos
candidatos = ["Alice", "Bruno", "Caio", "Duda"]

# Usamos o * para desempacotar a lista dentro do *args da função
sortear_premiado("Notebook", *candidatos)
```

## O que aconteceu aqui:

1. *args: A função aceita quantos nomes você mandar.

2. *candidatos: Você desempacotou sua lista de uma vez.

3. raise: Você criou sua própria exceção para garantir a regra do sorteio.

4. random.choice: Escolheu o item do intervalo.
