Chegou a hora! Vamos unir Herança, Encapsulamento, @property e Polimorfismo.

Criei este código base para você. O seu objetivo é completar o método atacar de forma polimórfica:

```python
class Personagem:
    def __init__(self, nome, vida):
        self.nome = nome
        self.__vida = vida # Encapsulado

    @property
    def vida(self):
        return self.__vida

    def atacar(self):
        pass # Será sobrescrito (Polimorfismo)

class Guerreiro(Personagem):
    def atacar(self):
        return f"{self.nome} golpeia com sua espada!"

class Mago(Personagem):
    def atacar(self):
        return f"{self.nome} lança uma bola de fogo!"

# --- SEU DESAFIO ---
# 1. Crie uma lista com um Guerreiro e um Mago.
# 2. Use um loop para fazer ambos atacarem.
# 3. Adicione uma classe 'Arqueiro' que também herda de Personagem.
```
