```python
livros = ["A Arte de Aprender", "Castelo Interior", "Divina Comédia"]

def inserir_dados():
    input_inserir = input("Digite o nome do livro que deseja inserir: ").title()
    livros.append(input_inserir)

    return f"O Lívro '{input_inserir} foi cadastrado.'", print(livros)


```

```python
def menu_cadastrar():
    
    while True:
        print("""
              >>> MENU DE CADASTRO DE LIVROS <<<

              Selecione umas das opções disponíveis:
              
              1 - Cadastrar um novo livro;
              2 - Registro de livros cadastrados;
              3 - Retornar para o menu principal.

              """)
        
        opcoes_de_cadastrar = input("Digite o número desejado ")

        if opcoes_de_cadastrar == "1":
            inserir_dados()

        elif opcoes_de_cadastrar == "2":
            continue
        
        elif opcoes_de_cadastrar == "3":
            continue
```

# Treinando com .append()

```python

def registrar_na_lista():
    lista = []
    count = 5

    for numero in range(count):
        adiciona = input("Digite algo")
        lista.append(adiciona)
        count -= 1
        print(lista)

    return print("\nLista final: ", lista)

registrar_na_lista()
```

# Treinando com .remove()

```python
lista_completa = [1, 2, 3, 4, 5]

def remove_valores():
    valor_removido = int(input(f"A lista possuí os valore {lista_completa}. Digite o valor que quer remover: "))
    lista_completa.remove(valor_removido)
    print(lista_completa)

remove_valores()
```

```python
# v2

def remove_valores():
    lista_completa = [1, 2, 3, 4, 5]
    contador = 3

    for valor in range(contador):
        valor_removido = int(input(f"A lista possuí os valore {lista_completa}. Digite o valor que quer remover: "))
        lista_completa.remove(valor_removido)
        contador -= 1
        print(lista_completa)

    return print("\nLista final: ", lista_completa)

remove_valores()
```

# Treinando com ambos append() e remove()

```python
input_menu = input("""Digite uma das opções: 1 - Adicionar livro e 2 Remover livro
                   Lista atual: {lista_livros}
                   """)

lista_livros = ["A Arte"]

def adicionar_livro():
    """ Adicionará um livro por vez na lista """

    input_adicionar = input(f"""Digite o nome do livro que quer ADICIONAR: """)
    lista_livros.append(input_adicionar)

    # Confirma que o livro foi adicionado e mostra a lista novamente
    print(f"""
        O Livro '{input_adicionar}' foi adicionado com sucesso.
        Lista atual: {lista_livros}
    """)

def remover_livros():
    """ Removerá um livro por vez na lista """

    input_remover = input(f"""Digite o nome do livro que deseja REMOVER: """)
    lista_livros.remove(input_remover)

    # Confirma que o livro foi removido e mostra a lista novamente
    print(f"""
        O Livro '{input_remover}' foi adicionado com sucesso.
        Lista atual: {lista_livros}
    """)



if input_menu == '1':
    adicionar_livro()

elif input_menu == '2':
    remover_livros()
else:
    print("fim")
```
