# ***append( )***-> Anexar elementos

Anexar (tradução direta de 'append') ou adicionar (não seria o termo correto, mas para ajudar em uma rápida compreensão do que ele faz) novos elemento/valores para dentro de uma lista; 

Inclusive até mesmo novas listas dentro de listas, dessa forma crinado o que chamamos de 'matrizes', como explicado na cartilha 01_01-listas.ipynb. 


***Atenção:*** O **append()** adiciona um novo valor por vez. Diferente do método **'extend()'**, que falaremos mais tarde sobre.

```python
lista = []

lista.append(1)
lista.append("Python")
lista.append("DIO.me")
lista.append([40, 30, 20]) # Adicona até mesmo listas
```

Conferindo os valores adicionados:

```python
print(lista)
```

## append( ) de listas dentro de listas:

```python
# Anexando uma lista dentro de outra, utilizando o método append()
lista_de_numeros = [1,2,3,4,5]
print(lista_de_numeros)

lista_de_mais_numeros = [6,7,8,9,10]

lista_de_numeros.append(lista_de_mais_numeros)
print(lista_de_numeros)
```

```python
print(lista)

for i in lista:
    print(i)
```

## ***"Atribuição por Índice" ou "Indexação com Atribuição"***

```python
lista[0] = 10

print(lista)
```

No trecho lista[0] = 10, você está atribuindo (ou modificando) um valor em uma posição específica da lista, acessando-a diretamente pelo índice.

Isso é chamado de atribuição por índice ou simplesmente indexação com atribuição.

A indexação altera um elemento já existente em uma posição fixa, enquanto append() adiciona um novo elemento sempre no final da lista.

# ***copy( )*** -> Copia a lista inteira

A lista é um objeto mutavél, ou seja, ela pode ser alterada. Com esse método, é possível criar uma nova lista, tendo uma nova instância (portanto não é a mesma lista que foi criada).

```python
# Trazendo novos valores a lista que foi limpa
lista.append(2)
lista.append("Felipe")
lista.append("Magalhães")
lista.append([31, 32, 33]) # Adicona até mesmo listas

print(lista)
```

```python
# Fazendo a cópia de uma lista, enquanto cria e atribui a uma outra:
lista2 = lista.copy()

# Comprovação que as listas são diferentes, embora uma é uma cópia, de todos os dados da outra
print(id(lista), lista)
print(id(lista2), lista2)
```

# ***count( )*** -> Contar quantas vezes um elemento aparece na lista

Método responsável por fazer a contagem de quantas vezes um valor repetido aparece em uma lista.

***Aplicação:*** este método necessita de um argument e, nele coloque o valor que deseja que ele encontre e conte quantas vezes este valor está na lista.

```python
cores = ["Vermelho", "Azul", "Amarelo", "Azul", "Vermelho", "Azul"]


print("Vermelho:", cores.count("Vermelho"))
print("Azul:", cores.count("Azul"))
print("Amarelo:", cores.count("Amarelo"))
```

# ***extend( )*** -> 'Extender' a lista anexando 1 ou mais elementos novos a ela.

Método muito utilizado. Neste é possível adicionar vários valores dentro de uma lista, inclusive pode adicionar os valores de uma lista em uma outra lista. 

***Aplicação:*** Este método necessita que dentro do seu argumento também contenham colchetes [] e dentro deste os valores que você quer.

***Obs.:*** 
- Diferente de append() que é um por vez.
- Ele não elimina valores duplicados

```python
linguagens = ["Python", "JavaScript", "C"]

print(linguagens)

linguagens.extend(["Java", "Rubi"])
print(linguagens)
```

```python
# adicionando valores de uma lista em outra lista:
linguagens_baixo_nivel = [ "C++" , "Rust" ]

linguagens.extend(linguagens_baixo_nivel)

print(linguagens)
```

'Extendendo' com valores duplicados:

```python
linguagens_baixo_nivel.extend(["C++", "Rust"])

print(linguagens_baixo_nivel)
```

# ***pop( )*** -> Remove um elemento por vez 

Uma de duas formas para remover um dado de uma lista. Ao usa-lo sem uma argumentação, ele remove o último dado em uma lista. Uma vez definido dentro de seu argumento um index, ele excluirá este index definido da lista

```python
linguagens.pop()

print(linguagens)
```

## Método pop( ) com o uso de index em seu parâmetro:

Remove e retorna o elemento de um índice específico

```python
linguagens.pop(1)

print(linguagens)
```

```python
lista_nova = [1, 2, 3]

removido = lista_nova.pop(1)  # Remove posição 1, retorna 2

print(lista_nova)     # [1, 3]

print(removido)  # 2
```

# ***sort( )*** -> Ordena a lista.

É um método das listas que ordena in-place (modifica a lista original). A lista que for string, retornará em ordem alfabética (por padrão) e se for número em ordem crescente (por padrão).

***Aplicação:***
- Se não houver argumento, ele ordenará por ordem alfabética;
- Se houver dentro do argumento 'reverse=True', será por ordem alfabética, mas será o contrário;
- Se houver dentro do argumento 'key=lambda x: len(x)', será por ordem do tamanho sendo crescente;
- Se houver dentro do argumento 'key=lambda x: len(x), reverse=True', será por ordem do tamanho; sendo decrescente.

```python
linguagens.sort()
print(linguagens)
```

```python
linguagens.sort(reverse=True)
print(linguagens)
```

```python
linguagens.sort(key=lambda x: len(x))
print(linguagens)
```

```python
linguagens.sort(key=lambda x: len(x), reverse=True)
print(linguagens)
```

## No caso de números:

```python
lista_de_numeros_nova = [1,7,9,2]

lista_de_numeros_nova.sort()
```

```python
print(lista_de_numeros_nova)
```

Perceba que o método sort( ) altera a lista original.

No caso de você querer ordernar essa lista, mas não altera-lá, utilize a função 'sorted()'.

Este é o próximo método que será apresentado:

# ***sorted()***

Função que faz o mesmo que o método sort( ), mas a sua forma de usar é um pouco diferente:

- Como é uma função, a lista que estiver dentro de seus parâmetros será ordenada;

- Você pode atribuir essa ação dentro de uma variável, o que a tornará uma nova lista.

```python
print(linguagens)
print(id(linguagens), linguagens) # analizando a id desta lista


# Atribuindo essa ação dentro de uma variável, o que a tornará uma nova lista:
lista_nova_ordenada = sorted(linguagens) # Melhor forma de usar a função é a coloca-lo em uma nova lista

print(id(lista_nova_ordenada), lista_nova_ordenada) # comprovando que ele é diferente por meio de sua id
```

```python
print(linguagens)

sorted(linguagens, key=lambda x: len(x))
```

```python
print(linguagens)

sorted(linguagens, key=lambda x: len(x), reverse=True)
```
