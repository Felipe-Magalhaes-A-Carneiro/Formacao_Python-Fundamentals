# ***Funcões que são compatíveis com Listas***



# ***len()***

Função que retorna o tamanho de uma lista. 

***Obs.:*** Aqui é feita utilizando o número de quantos index a lista selecionada possuí.

```python
linguagens = ["Python", "JavaScript", "C"]
```

```python
print(linguagens)
len(linguagens)
```

----------------------------------------------------------------------------------------------------

# ***range()***

Função que retorna um conjunto de valores que são delimitados pelos números em seus argumentos.

Se colocar 5, virá 1, 2, 3, 4.

Perceba que ele tem o mesmo modo de agir que um slicing, aonde este ignora o último número que é definido no argumento.

Está função funciona melhor quando acompanhado de classes como 'list' e o uso do 'for'

```python
list(range(5))
```

```python
list(range(5, 10))
```

----------------------------------------------------------------------------------------------------

# ***max()***

Retorna o maior valor em uma lista de números ou strings.

```python

numeros = [5, 2, 8, 1]
print(max(numeros))  # Saída: 8
```

----------------------------------------------------------------------------------------------------

# ***min()***

Retorna o menor valor em uma lista de números ou strings.

```python
print(min(numeros))  # Saída: 1
```

----------------------------------------------------------------------------------------------------

# ***Sorted()***

Função que faz o mesmo que o método sort(), mas a sua forma de usar é um pouco diferente:

é uma função embutida que aceita qualquer iterável, cria e ***retorna*** uma ***nova lista ordenada***, mantendo a original intacta.

```python
print(linguagens)

sorted(linguagens)
```

```python
print(linguagens)

sorted(linguagens, key=lambda x: len(x))
```

```python
print(linguagens)

sorted(linguagens, key=lambda x: len(x), reverse=True)
```

outro exemplo:

```python
a = [2,1,3]
a.sort()

print(f"A lista a = {a} foi ordernada, alterando assim a posição dos números")


b = [2,1,3]
c = sorted(b) # a possui o valor [2,1,3] e b possui o valor [1,2,3]

print(f"A lista b = {b} foi copiada e esta copia foi ordenada, a original ficou intacta.")
print(f"Já a cópia c é {c} que é a lista b sendo ordenada pela função sorted()")
```

# ***tuple()***

A função tuple(), transforma uma lista em uma tupla

```python
dados_tupla = (4, 5, 6)

# Mudando de TUPLA para LISTA:
list(dados_tupla)
```

----------------------------------------------------------------------------------------------------

# ***tuple() depois de list()***

A função tuple() também pode ser utilizada em uma lista, porém aqui trazemos ela para mostrar que é possível desfazer a tasformação e Tupla em Lista, utilizando esta função tuple()

```python
# Retornando os seus tipos:
tuple(list(dados_tupla))
```
