# ***Criando Tuplas***

Tuplas são estruturas de dados muito parecdias com as listas, a principal diferença é que tuplas **são IMUTÁVEIS** enquanto listas são mutaveis. Podemos criar tuplas através da classe **tuple**, ou colocando valores separados por vírgula de parenteses.

Portanto, Listas e Tuplas são praticamente a mesma coisa, porém o fator que muda isso é: que Listas são mutaveis e Tuplas são Imutaveis, ambas permitem que seus dados sejam acessados, mas apenas uma NUNCA permitirá alteração (Tuplas).

***Obs nº1.:*** Quando é recomendado o uso de Tuplas então? A Tupla é recomendado para momentos no código que você precisa ter certaza que aqueles dados dentro dela, poderão ser acessados mas NUNCA serão alterados. O que traz uma segurança para o programador. Suponhamos que em uma dada situação em que você esteja em grupo de trabalho, alguém mude o valor de algo que não deveria em uma Lista, prejudicando o código. Tuplas não permitiriam que essa situação ocorrece.

***Obs nº2.:*** No Formato 2 de criação de tuplas (declarar o nome, sinal de atribuição, e parênteses e os valores dentro deles), nas boas práticas, é recomendado que você adicione mais uma vírgula depois de ter declarado um dado. O motivo é que parênteses também é usado para operações matemáticas para indicar precendência, portanto o interpretador do Python pode se confundir quando você estiver criando uma tupla e ele achar que é uma precendência de números para um cálculo.

```python
# Formato 1 de criação:

letras = tuple("Python")

print(letras)
```

```python
# Formato 2 de criação:

letras_java = ("J","a","v","a",) # Como recomendado em obs., colocamos uma ' , ' no final como boas práticas na criação de tuplas
print(letras_java)
```

# ***Acesso Direto - Tuplas***

Exatamente como em listas, como a tupla é uma sequência, podemos acessar seus dados utilizando índices. Contamos o índice de determinada sequência apartir do zero, assim como nas listas.

```python
letras_java[0]
```

```python
letras_java[-1]
```

# ***Tuplas Aninhadas***

Tuplas podem armazenar todos os tipos de objeto Python, portanto podemos ter tuplas que armazenam outras tuplas. Com isso podemos criar estruturas bidimensionais (matrizes/tabelas), e acessar informando os índices de linha e coluna.

```python
matriz_tupla = (
    (1,"a",3),
    (2,"b",2),
    (3,"c",1),
)

matriz_tupla[0]
```

```python
matriz_tupla[0][2]
```

```python
matriz_tupla[-1]
```

```python
matriz_tupla[-1][-1]
```

# ***Em resumo:***

Por que eu usaria ***Tuplas*** em vez de usar as ***Listas***?

Tuplas, como já dito, são imutaveis, diferente das listas. Portanto, muito recomendado quando você quer armazenar dados e que estes não sejam aleterados.

# ***Fatiamento***

Além de acessar elementos diretamente, podemos extrair um conjunto de valores de uma sequência. Para isso basta passar o índice e/ou final para acessar o conjunto. Podemos ainda informar quantas posições o cursor deve "pular" no acesso.

***Obs.:*** Assim como em listas, o número depois de ' : ' é desconsiderado, atrazendo apenas o que vem antes.

```python
letras_python = ('P', 'y', 't', 'h', 'o', 'n')

letras_python[3:]
```

```python
letras_python[:3] # Perceba que, assim como em listas, o número depois de ' : ' é desconsiderado, atrazendo apenas o que vem antes.
```

```python
letras_python[0:3]
```

```python
letras_python[0:5:2]
```

```python
letras_python[::]
```

```python
letras_python[::-1]
```

# ***Iterar uma Tupla***

A forma é a mesma feita em listas. Utilizamos *for* para fazê-lo.

```python
carros = ("Gol", "Palio", "Celta",)

for carro in carros:
    print(carro)
```

```python
for i, posicao in enumerate(carros):
    i += 1
    print(f"Marca número {i} é: {posicao}")
```

# ***Mudando de Tupla para Lista e vice-versa***

Embora exista a diferença entre lista ser mutavel e tuplas não, é possível alterar o seu tipo, utilizando a seguinte função:


```python
dados_lista = [1,2,3]

dados_tupla = (4,5,6)

# Mudando de LISTA para TUPLA:
tuple(dados_lista)
```

```python
# Mudando de TUPLA para LISTA:
list(dados_tupla)
```

```python
# Mudando de LISTA para TUPLA:
list(tuple(dados_lista))
```

```python
# Retornando os seus tipos:
tuple(list(dados_tupla))
```
