```python
# 1. Crie a função 'soma_tupla' para receber uma tupla de números inteiros:
def soma_tupla(tupla_numeros):
    # A função 'sum' já faz o trabalho de somar todos os elementos
    return sum(tupla_numeros)

if __name__ == "__main__":
    # Lê a entrada como uma string (ex: "10 20 30")
    entrada = input()
    
    # Converte a string em uma lista de strings cortando nos espaços
    lista_strings = entrada.split()
    
    # 2. Use a função "map()" para aplicar "int()" a cada elemento (conforme a instrução)
    # 3. Converta o resultado para uma tupla e armazene em "elementos"
    elementos = tuple(map(int, lista_strings))
    
    # Calcula o resultado chamando a função criada
    resultado = soma_tupla(elementos)
    
    # Exibe a saída no formato amigável solicitado
    print(f"A soma dos elementos da tupla é: {resultado}")
```

```python
def elementos_comuns(lista1, lista2):
    # 1. Converte os elementos para inteiros usando map
    # 2. Transforma em set (conjunto) para poder fazer a interseção
    set1 = set(map(int, lista1))
    set2 = set(map(int, lista2))
    
    # 3. Encontra a interseção e converte de volta para lista
    # Usamos sorted() caso queira a lista em ordem, ou apenas list()
    return sorted(list(set1.intersection(set2)))

# Leitura das listas
entrada1 = input().split()
entrada2 = input().split()

# Verifica se todos os elementos das listas são numéricos (isdigit)
if all(item.replace('-', '').isdigit() for item in entrada1) and \
   all(item.replace('-', '').isdigit() for item in entrada2):
    
    comuns = elementos_comuns(entrada1, entrada2)
    print(f"Elementos comuns às duas listas: {comuns}")
else:
    print("Entrada inválida.")
```
