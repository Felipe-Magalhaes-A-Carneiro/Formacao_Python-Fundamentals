# ***Funções utilizadas em Tuplas***



# ***list()***

A função list(), transforma uma tupla em uma lista

```python
dados_tupla = (4, 5, 6)

# Mudando de TUPLA para LISTA:
list(dados_tupla)
```

# ***len()***

Retorna o tamanho de uma tupla.

Também é usado em Listas

```python
len(dados_tupla)
```

# ***tuple() depois de list()***

A função tuple() também pode ser utilizada em uma lista, porém aqui trazemos ela para mostrar que é possível desfazer a tasformação e Tupla em Lista, utilizando esta função tuple()

```python
# Retornando os seus tipos:
tuple(list(dados_tupla))
```

# ***tuple() em strings e números***

É possível transformar strings em tuplas

```python
palvra = "Python"

tuple(palvra)
```

```python
n = 1,2,3,4

tuple(n)
```
