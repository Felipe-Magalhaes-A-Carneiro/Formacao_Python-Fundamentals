# ***For - Loop***

Nada mais do que repetir uma certa ação em um certo número de vezes.

sintaxe:

Para cada **disto** em/dentro de **repita tantas vezes**:
    
    - isso...

```python
for "variável" in range("esse número de vezes"):
    "Faça isso..."
```

- range() é comumente usado, mas podem ser usados outras funções, listas, tuplas, dicionários e etc;
- em 'isto' é a variável que será usada como base para ser repetida
    - A variável pode ser inclusive determinada ali

```python
for n in range(11):
    print(f"O 'n' é o nome que dou a cada número dentro deste loop, utilizando a função range() para alcançar cada número que coloquei nos seus argumentos. Logo n é {n}")
```

Em convenções de Python, os programadores constumam nomear o seu for de simples uso, de ' _ ' para loops não serão usados novamente, portanto que seu uso seja para alcançar alguma outra coisa de mais importantem logo ele foi um simples meio para chegar lá, ou seja, se repetirmos o exemplo a cima, ficaria:

for _ range (11):

Então fique atento para que, caso você veja isso em algum código, essa é a explicação para tal situação.


# ***Iterando Sobre Sequências***

Iterar ou 'passar' por cada elemento contido em um dado. Seja ele uma String, Lista, Tupla, Dicionários. 

Nem tudo é iterável. O tipo de dado int, float não são. Para poder faze-lo, O Python entende que o número 10 (tipo int) é ele e ponto. Já uma String de valor 'Python' por exemplo, possuí caracteres que de um a um formam o valor que está na String (P + y + t + h + o + n), logo iterável. 

```python
for letra in "Python":
    print(letra)
```

Para fazer com que o Python vá de 0 a 10 (pois como sabemos é assim que chegamos até o 10), por exemplo, você precisará da função range(), pois estamos informando ao Python que 'o percurso até chegar ao 10, traga cada um deles para mim, mas ignore o número depois dele' e aí sim ele irá trazer o número de repetições que você quer, caso o contrário, não será possível.

```python
for numero in range(11): # ou seja, ele irá iterar (percorrer) número por número
    print(numero)
```

Forçando o erro:

```python
for numero in 11: # ou seja, ele não consegue iterar (percorrer) número por número
    print(numero)
```

# ***Iterar Listas e Tuplas***

```python
lista_exemplo = ["Python", "Java", "JavaScript"]

for dado in lista_exemplo:
    print(dado)
```

```python
tupla_exemplo = (1, 2, 3, 4, 5, ) # Boas práticas colcoar um , no final de uma tupla

for valores in tupla_exemplo:
    print(f"Temos os valores: {valores}")
```

```python
lista_clientes = [ ("Felipe", "xxx", "felipe@gmail.com"), ("Ana", "yyy", "ana@gmail.com") ]

for cliente in lista_clientes:

    nome = cliente[0]
    cpf = cliente[1]
    email = cliente[2]

    print(f"""
          Cliente: {nome}
          CPF: {cpf}
          email: {email}
          """)
```

1. Existem duas camadas de dados:
    - A primeira: É a Tupla, que será acessado pelo variavel for 'cliente', ou seja, temos duas tuplas;
    - A segunda: É os valores dentro da tupla, que será acessado pelas variáveis nome, cpf e email, dentro da variável do for 'cliente'.

## ***Melhorando o código anterior***

```python
lista_clientes = [ ("Felipe", "xxx", "felipe@gmail.com"), ("Ana", "yyy", "ana@gmail.com") ]

for cliente in lista_clientes:

    nome, cpf, email = cliente

    print(f"""
          Cliente: {nome}
          CPF: {cpf}
          email: {email}
          """)
```

***Explicação:***
Este é outro módo de fazer o mesmo que no anterior, porém perceba que temos um uso mais limpo, rápido e elegível de entender o código.

Estamos dizendo para o Python, que já entendeu que os valores "Felipe", "xxx" e "felipe@gmail.com", por exemplo, são os index[0], index[1] e index[2] nesta mesma ordem; Portanto, basta que escrevamos as suas variáveis e declaremos elas na ordem adquada na variável for 'cliente' que intera-rá cada um dos index;

nome, cpf, email = cliente
