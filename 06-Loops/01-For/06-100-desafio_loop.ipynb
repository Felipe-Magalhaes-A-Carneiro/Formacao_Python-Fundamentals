# Desafios - Iteração

- Dado uma sequência de números, calcule a soma e média dos números;
    Atenção: Não vale usar a função sum().


```python
soma = 0
media = 0

while True:
    numero = input("Digite um número: ")

    if numero != 'sair':
        
        resultado_soma = soma + int(numero)
        soma += int(numero)
        media = resultado_soma / 2

        print(f"Soma de {numero} + {soma} é: {resultado_soma} a média é {media}")

    else:
        print("Saindo do programa...")
        break
```

```python
# outra forma de resolver:

numeros = [5, 3, 6, 10, 28]
soma = 0
media = 0

for n in numeros:
    soma += n
    media = soma / len(numeros)

print(f"A soma dos números é {soma} e a média é {media}")
```

- Dado uma sequência de números, calcule o maior valor da sequência;
    Atenção: Não vale usar a função max().

```python
valores = [53, 63, 96, 10, 28]
valor_comparado = valores[0]


for valor in valores:

    if valor > valor_comparado:
        maior_valor = valor
        
print(f"O maior valor nesta lista é {maior_valor}")
```

- Dado uma lista de palavras, printe todas as palavras com pelo menos 5 caracteres;

```python
lista_palavras = ["Python", "Casa", "Barco", "Solda", "abelha", "amor", "dor"]

for palavra in lista_palavras:
    if len(palavra) >= 5:
        print(f"A palavra {palavra} tem {len(palavra)} letras")
```
