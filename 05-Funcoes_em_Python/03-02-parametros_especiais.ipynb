# ***Parâmetros Especiais***

Os Parâmetros Especiais no Python são delimitadores que você coloca na definição da sua função (def) para forçar quem a utiliza a seguir certas regras de "etiqueta" na passagem de argumentos.

Eles servem basicamente para dizer: "A partir daqui, você só pode passar valores desse jeito".

Os dois principais delimitadores são a barra (/) e o asterisco (*).

----------------------------------------------------------------------------------------------------

## 1. Parâmetros Apenas Posicionais (/)
Ou seja sequênciais, são conhecido como **'parâmetros posicionais'**

Quando você coloca uma barra em uma função, todos os argumentos antes dela devem ser passados apenas pela posição, sem o nome da chave (ex: func(10) e não func(x=10)).

```python
def configurar_servidor(ip, porta, /):
    print(f"Servidor em {ip}:{porta}")

# CORRETO:
configurar_servidor("192.168.0.1", 8080)

# ERRO:
# configurar_servidor(ip="192.168.0.1", porta=8080)
```

Por que usar? Serve para quando os nomes dos parâmetros não importam ou podem mudar no futuro, e você não quer que o usuário se apegue a eles.

----------------------------------------------------------------------------------------------------

## 2. Parâmetros Apenas Nomeados (*)
Conhecidos como **'parâmetros nomeados'**

O asterisco isolado obriga que tudo o que vier depois dele seja passado obrigatoriamente com o nome da chave (chave=valor).

```python
def enviar_email(destinatario, *, assunto, corpo):
    print(f"Enviando '{assunto}' para {destinatario}")

# CORRETO:
enviar_email("ana@email.com", assunto="Olá", corpo="Tudo bem?")

# ERRO:
# enviar_email("ana@email.com", "Olá", "Tudo bem?")
```

Por que usar? Isso evita erros graves. Imagine uma função deletar_usuario(id, confirmar=False). Se você esquecer o nome da chave, pode acabar deletando alguém sem querer. O * força a clareza.

----------------------------------------------------------------------------------------------------

## 3. O Mapa Completo

Você pode combinar tudo em uma única função. A ordem no Python segue este padrão:

$$def\ nome(posicional\_apenas, /, padrão, *, nomeado\_apenas):$$

```python
def complexa(pos, /, padrao, *, nomeado):
    print(pos, padrao, nomeado)

# 'pos' deve ser valor puro
# 'padrao' pode ser qualquer um
# 'nomeado' DEVE ter o nome
complexa(1, 2, nomeado=3)
complexa(1, padrao=2, nomeado=3)
```

Caso você não faça como delimitou na criação da função e seus parâmetros, será dado o seguinte erro:

```python
def complexa(pos, /, padrao, *, nomeado):
    print(pos, padrao, nomeado)

# 'nomeado' DEVE ter o nome, então vamos forçar um erro fazendo diferente do exemplo acima, Dessa forma:
complexa(1, 2, 3) # FORMA ERRADA DE FAZER, NESTE EXEMPLO
```

Ele entende os parâmetros anteriores, mas desconsidera o terceiro por ter sido usado de forma errada

## 4. Por que isso é importante na Programação Orientada a Objetos?

Ao criar Classes, você muitas vezes quer que os métodos sejam muito claros.

Se você está criando um método transferir(valor, /, *, destino), você garante que o programador sempre escreva destino=conta_b, tornando o código muito mais legível ("auditável").

Resumo Visual:

* / (Barra): Olha para a esquerda. Tudo atrás de mim é apenas posição.

* (Asterisco): Olha para a direita. Tudo na minha frente exige um nome.
