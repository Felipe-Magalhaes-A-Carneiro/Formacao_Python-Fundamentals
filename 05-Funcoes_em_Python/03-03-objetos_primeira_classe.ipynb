# ***Objetos de primeira classe***

Em Python, dizemos que "Funções são objetos de primeira classe" (ou First-Class Citizens). Isso significa que o Python trata suas funções com o mesmo respeito e liberdade que trata um número inteiro ou uma string.

Para algo ser considerado um "Cidadão de Primeira Classe", ele precisa ser capaz de fazer três coisas:

----------------------------------------------------------------------------------------------------

## 1. Ser atribuido a uma variável

Você pode pegar uma função e "guardá-la" dentro de um nome, como se fosse um valor.

```python
def gritar(texto):
    return texto.upper() + "!!!"

# Atribuindo a função a uma variável (sem os parênteses!)
falar_alto = gritar 

print(falar_alto("olá")) # Saída: OLÁ!!!
```

----------------------------------------------------------------------------------------------------

## 2. Ser passado como argumento para outra função

Isso é o que permite a existência de coisas incríveis como o método .sort(). Você pode enviar uma "lógica" para dentro de outra função.

```python
def aplicar_operacao(lista, funcao):
    return [funcao(item) for item in lista]

numeros = [1, 2, 3]
# Passamos a função 'str' (nativa) como se fosse um dado
print(aplicar_operacao(numeros, str)) # Saída: ['1', '2', '3']
```

----------------------------------------------------------------------------------------------------

## 3. Ser retornado por outra função

Uma função pode "fabricar" e devolver outra função. Isso é a base de conceitos avançados como Decoradores e Closures.

```python
def saudacao(tipo):
    def bom_dia(nome):
        return f"Bom dia, {nome}!"
    def boa_noite(nome):
        return f"Boa noite, {nome}!"

    if tipo == "manha":
        return bom_dia
    else:
        return boa_noite

# 'f' agora é a função 'bom_dia'
nova_saudacao = saudacao("manha")
print(nova_saudacao("Lucas"))
```

----------------------------------------------------------------------------------------------------

## Por que isso importa na Orientação a Objetos?

Nas Classes nós temos os Métodos. Como as funções são objetos de primeira classe, você pode:

1. Armazenar métodos em listas.

2. Trocar o comportamento de um objeto em tempo de execução apenas substituindo a função que ele carrega.

3. Criar funções que recebem objetos e chamam seus métodos dinamicamente.

Outro exemplo mais claro:

```python
def somar(a, b): # crio uma função que soma dois valores
    return a + b

# crio outra função que mostrará o resultado com texto editado, usando a função criada anteriormente como um parametro:

def exibir_resultado(a, b, funcao_escolhida):
    resultado = funcao_escolhida(a, b)
    print(f"A soma dos números {a} + {b} é {resultado}")

exibir_resultado(10, 10, somar) # a função 'somar', é usado como um parametro

# Tudo isso é possível porque funções são objetos, e não só isso 'objetos de primeira classse', 
```

# ***Escopo local e escopo global***

-----------------------------------------------------
***ESSA NÃO É UMA BOA PRÁTICA E DEVE SER EVITADA.***
-----------------------------------------------------
Python trabalha com escopo local e global, dentro do bloco da função o escopo é local. Portanto alterações ali feitas em objetos imutáveis serão perdidas quando o método terminar de ser executado. Para usar objetos globais utilizamos a palavra-chave 'global', que informa ao interpretador que a varipavel que está sendo manipulada no escopo local é global.

```python
salario = 2000

def salario_bonus(bonus):
    global salario
    salario += bonus
    return salario

salario_bonus(500)
```
