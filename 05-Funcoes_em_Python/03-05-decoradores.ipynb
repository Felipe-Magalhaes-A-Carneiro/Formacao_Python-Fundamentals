# ***@decoradores***

----------------------------------------------------------------------------------------------------

Os Decoradores (ou Decorators) são uma das funcionalidades mais elegantes e poderosas do Python. Eles permitem que você "embrulhe" uma função dentro de outra para adicionar comportamentos a ela sem modificar o código original da função.

Lembra que vimos que as funções são objetos de primeira classe? Os decoradores só existem porque podemos passar funções como argumentos e retornar funções como resultado.

----------------------------------------------------------------------------------------------------

## ***1. A Anatomia de um Decorador***

Um decorador é, essencialmente, uma função que recebe outra função e devolve uma versão "turbinada" dela.

```python
def meu_decorador(funcao_original):
    def wrapper():
        print("Antes da função ser executada...")
        funcao_original()
        print("Depois da função ser executada!")
    return wrapper

@meu_decorador
def oi():
    print("Olá, mundo!")

oi()
```

O símbolo @meu_decorador (conhecido como "Pie Syntax") é apenas um atalho para:

oi = meu_decorador(oi)

----------------------------------------------------------------------------------------------------

## ***2. Por que usar Decoradores?***

Eles são perfeitos para o princípio ***DRY*** (Don't Repeat Yourself). Imagine que você tem 10 funções e quer medir quanto tempo cada uma leva para rodar. Em vez de escrever o código de cronômetro em todas as 10, você cria um decorador.

Exemplo: Decorador de Cronômetro

```python
import time

def cronometro(funcao):
    def wrapper(*args, **kwargs): # Aceita qualquer argumento
        inicio = time.time()
        resultado = funcao(*args, **kwargs)
        fim = time.time()
        print(f"A função '{funcao.__name__}' levou {fim - inicio:.4f} segundos.")
        return resultado
    return wrapper

@cronometro
def conta_demorada(n):
    sum(range(n))

conta_demorada(10000000)
```

----------------------------------------------------------------------------------------------------

## ***3. Decoradores com Argumentos (* args e ** kwargs)***

Note que no exemplo acima usei *args e **kwargs dentro do wrapper. Isso é fundamental! Se a função original recebe argumentos e o seu decorador não estiver preparado para repassá-los, o código vai quebrar.

----------------------------------------------------------------------------------------------------

## ***4. Decoradores Famosos que já vêm no Python***

Você não precisa apenas criar os seus; o Python usa muitos decoradores nativos, especialmente na Orientação a Objetos:

- @property: Transforma um método em um "atributo" (você acessa sem os parênteses).

- @staticmethod: Define um método que não precisa do self.

- @classmethod: Define um método que recebe a classe (cls) em vez da instância.

----------------------------------------------------------------------------------------------------

## ***5. Analogia do Presente***

Imagine que você tem um presente (a função original).

1. Você coloca o presente dentro de uma caixa (o wrapper).

2. Você coloca um laço e um cartão na caixa (o comportamento extra).

3. Quem recebe o presente ainda recebe o que estava lá dentro, mas agora com uma embalagem especial.

----------------------------------------------------------------------------------------------------

# ***Resumo:***

- Decoradores servem para modificar o comportamento de funções ou classes de forma limpa.

- Eles usam o símbolo @ acima da definição da função.

- São ideais para: Logging (registro de logs), Autenticação, Cache e Medição de tempo.
