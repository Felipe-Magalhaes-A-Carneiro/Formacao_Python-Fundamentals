# ***Funções*** *"Programar de maneira estruturada".*

Aqui estamos falando do segundo tipo de paradigma: "Paradigma Convencional" (Criação de códigos utilziando funções)

É uma evolução do primeiro paradigma: "Paradigma Imperativo"(Criação de códigos por "leitura e execução de linha por linha")

Se o Python é uma caixa de ferramentas, as funções são as suas máquinas personalizadas.

Elas servem para empacotar um bloco de código que você pretende usar várias vezes, evitando que você tenha que "copiar e colar" as mesmas instruções.

## 1. O que é uma Função?

Uma função é um bloco de código que só é executado quando é chamado. Você pode enviar dados para ela (parâmetros) e ela pode te devolver dados (retorno).

Pense em uma máquina de café:

1. Você coloca o grão (Entrada/Parâmetro).

2. Ela processa internamente (Corpo da função).

3. Ela te entrega o café pronto (Saída/Retorno).

## 2. A Estrutura de uma Função (def)

Para criar uma função, usamos a palavra reservada def ("défh"), de *'define'*.

```python
def calcular_desconto(preco, percentual): # 1. Definição e Parâmetros
    desconto = preco * (percentual / 100) # 2. Lógica interna
    valor_final = preco - desconto
    return valor_final                   # 3. Resultado final

calcular_desconto(100, 20)
```

Anatomia da função:
- def: Avisa ao Python: "Vou criar uma ferramenta agora!".

- Nome da função: Como você a chamará depois (ex: calcular_desconto).

- Parâmetros (): Os "ingredientes" que a função precisa para trabalhar.

- return: A palavra mágica que faz a função trazer o resultado para fora. Se não houver return, a função executa a lógica mas não devolve nada para ser guardado em uma variável.

----------------------------------------------------------------------------------------------------

## 3. Chamando a Função

Depois de definida, você a usa chamando pelo nome e passando os valores nos parênteses

```python
novo_preco = calcular_desconto(100, 10) # 100 reais com 10% de desconto

print(novo_preco) # Saída: 90.0
```

----------------------------------------------------------------------------------------------------

## 4. Por que usamos funções?

1. Reutilização: Escreva uma vez, use mil vezes.

2. Organização: Se você tiver um erro no cálculo de desconto, só precisa consertar em um lugar, em vez de procurar em todo o seu código.

3. Abstração: Você não precisa saber como a função funciona por dentro para usá-la. (Como o print() ou o len(), que são funções prontas do Python).

# ***Args e Kwargs***

Os nomes *args e **kwargs podem parecer estranhos no começo, mas eles são apenas "coringas" que permitem que suas funções sejam extremamente flexíveis.

***Eles servem para quando você não sabe quantos argumentos alguém vai passar para a sua função.***

## ***1. O args (Argumentos Posicionais)***

O *args permite que você passe uma quantidade ilimitada de valores soltos. O Python pega todos ***esses valores e os empacota em uma Tupla***.

- O símbolo é o que importa: O nome "args" é apenas uma convenção. O segredo é o asterisco único (*).

- Como funciona: Imagine uma função que soma números. Com *args, ela pode somar 2 números ou 200.

```python
def somar_tudo(*args):
    # args é uma tupla: (1, 2, 3, 4)
    return sum(args)

print(somar_tudo(10, 20))         # Resultado: 30
print(somar_tudo(1, 2, 3, 4, 5))  # Resultado: 15
```

## ***2. O kwargs (Argumentos Nomeados)***

O **kwargs funciona de forma semelhante, mas para argumentos que têm um nome (como chave=valor). O Python empacota esses nomes e valores em um Dicionário.

- O símbolo: Dois asteriscos (**).

- Como funciona: É útil para passar configurações ou dados opcionais rotulados.

```python
def exibir_perfil(**kwargs):
    # kwargs é um dicionário: {'nome': 'Ana', 'cargo': 'Dev'}
    for chave, valor in kwargs.items():
        print(f"{chave}: {valor}")

exibir_perfil(nome="Ana", cargo="Dev", nivel="Sênior")
```

----------------------------------------------------------------------------------------------------

## ***3. Resumo da Diferença***

Imagine que você está enviando uma caixa pelo correio:

- *args: Você joga os itens soltos na caixa. A ordem em que você os coloca é o que define a posição deles.

- **kwargs: Você coloca cada item com uma etiqueta (nome). Não importa a ordem, você vai achar o item pelo nome na etiqueta.

----------------------------------------------------------------------------------------------------

# ***4. A Ordem Correta***

Se você for usar todos os tipos de argumentos em uma única função, o Python exige uma ordem específica para não se confundir:

1. Argumentos padrão (obrigatórios).

2. *args (extras soltos).

3. **kwargs (extras nomeados).

```python
def super_funcao(nome, *args, **kwargs):
    print(f"Olá {nome}")
    print(f"Args recebidos: {args}")
    print(f"Kwargs recebidos: {kwargs}")

super_funcao("Lucas", 1, 2, 3, cor="azul", ativo=True)
```

## ***Por que isso é útil?***

Muitas funções famosas do Python (como o print()) usam isso. O print aceita quantos itens você quiser (*args) e também aceita argumentos nomeados como sep="-" ou end="\n" (**kwargs).

## outro exemplo feito pelo professor:

```python
def exibir_poema(data_extenso, *args, **kwargs):

    texto = "\n".join(args)

    
    meta_dados = "\n".join([f"{chave.title()}: {valor}" for chave, valor in kwargs.items()])

    # Edição da mensagem e como ela será apresentada:
    mensagem = f"{data_extenso}\n\n{texto}\n\n{meta_dados}"

    print(mensagem)

exibir_poema("Zen of Python", "Beautiful is better than ugly.", autor = "Tim Peters", ano = 1999)
```

```python
def salvar_poema(titulo_arquivo, data_extenso, *args, **kwargs):
    texto = "\n".join(args)
    meta_dados = "\n".join([f"{chave.title()}: {valor}" for chave, valor in kwargs.items()])
    
    mensagem = f"{data_extenso}\n\n{texto}\n\n{meta_dados}"

    # Criando e escrevendo no arquivo
    nome_final = f"{titulo_arquivo}.txt"
    with open(nome_final, "w", encoding="utf-8") as arquivo:
        arquivo.write(mensagem)
    
    print(f"Poema salvo com sucesso em: {nome_final}")

# Chamada da função
salvar_poema("zen_python", "Zen of Python", "Beautiful is better than ugly.", autor="Tim Peters", ano=1999)
```

# ***unpacking - "desempacotamento"***

O "desempacotamento" (ou unpacking) é o processo inverso do que vimos antes. Em vez de a função juntar vários itens em uma caixa, você já tem uma caixa (lista ou dicionário) e quer espalhar o conteúdo dela nos parâmetros da função.

## 1. Desempacotando Listas e Tuplas com *

Imagine que você tem uma lista com três números e uma função que recebe três parâmetros. Em vez de passar *lista[0], lista[1], lista[2]*, você usa o asterisco.

```python
def calcular_volume(comprimento, largura, altura):
    return comprimento * largura * altura

dimensoes = [10, 5, 2]

# O asterisco "explode" a lista e passa cada item para um parâmetro
resultado = calcular_volume(*dimensoes)

print(resultado) # Saída: 100
```

## 2. Desempacotando Dicionários com **
Isso é muito comum em sistemas de configurações. Se você tem um dicionário onde as chaves têm exatamente os mesmos nomes dos parâmetros da função, o ** faz a mágica acontecer.

```python
def criar_usuario(nome, idade, cargo):
    print(f"Salvando: {nome}, {cargo}, {idade} anos.")

dados_da_web = {
    "nome": "Carla",
    "idade": 30,
    "cargo": "Engenheira"
}

# Os dois asteriscos dizem: "Use as chaves do dict como nomes de parâmetros"
criar_usuario(**dados_da_web)
```

## 3. Por que isso é útil?

- Limpeza: O código fica muito mais curto e fácil de ler.

- Integração: Se você recebe dados de um banco de dados ou de uma API (que geralmente vêm como dicionários), você pode passá-los direto para suas funções sem precisar mapear um por um.

```python
import random

def sortear_premiado(premio, *nomes):
    try:
        if len(nomes) < 2:
            raise ValueError("É necessário pelo menos 2 nomes para o sorteio!")
            
        vencedor = random.choice(nomes)
        print(f"O prêmio '{premio}' vai para: {vencedor}!")
        
    except ValueError as erro:
        print(f"Erro no sorteio: {erro}")

# Temos uma lista de candidatos
candidatos = ["Alice", "Bruno", "Caio", "Duda"]

# Usamos o * para desempacotar a lista dentro do *args da função
sortear_premiado("Notebook", *candidatos)
```

## O que aconteceu aqui:

1. *args: A função aceita quantos nomes você mandar.

2. *candidatos: Você desempacotou sua lista de uma vez.

3. raise: Você criou sua própria exceção para garantir a regra do sorteio.

4. random.choice: Escolheu o item do intervalo.
