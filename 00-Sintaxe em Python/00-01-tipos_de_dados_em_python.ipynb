# ***Tipos de Dados***

----------------------------------------------------------------------------------------------------

Em programação, tipos de datas são um conceito importante.

Variáveis podem armazenar diferentes tipos de data, e diferentes tipos de data podem fazer coisas diferentes...

Python possuí os seguintes tipos de data built-in (feitos para estarem nele por padrão), nesta categoria existem:

| Descrição do dado | Tipo de dado |
| :--- | ---: |
| Tipo Texto: | `str` |
| Tipo Númerico | `int`, `float`, `complex` |
| Tipo Sequência | `list`, `tuple`, `range` |
| Tipo Mapeamento | `dict` |
| Tipo Conjunto (set) | `set`, `rozenset` |
| Tipo Booleano | `bool` |
| Tipo Binário | `bytes`, `bytearray`, `emoryview` |
| Tipo Nenhum (None) | `NoneType` |


----------------------------------------------------------------------------------------------------

### ***type( )***

Se você quiser descobrir qual é o tipo de dado que uma variável ou valor possui, utilize a função 'type()' e, dentro de seus parâmetros, informe a variável/valor que quer que ele revele qual é o seu tipo.

```python
x = 5
```

Vamos descobrir qual é o tipo de dado que 'x' esta armazenando:

```python
type(x)
```

Correto. Como vimos na tabela a cima, tipos de dados numéricos são: `int`, `float`, `complex`; neste caso, por ser um valor inteiro '5', então é `int`.
