# ***input()***

----------------------------------------------------------------------------------------------------

Uma função que permite que o usuário possa interagir com o programa. Ele escreve, o Python pode (ou não) armazenar o que foi escrito em uma variável.

O que nos permite tratar essa variavel depois e fazermos vários tipos de operação com ela. As possíbilidades são várias e será adequado o seu uso segundo a vontade do programador.

```python
input()
```

Perceba que aqui, eu não armazenei ela em nenhuma variável e nem escrevi nada em seus parâmetros / argumentos (Nome dado para o que é aceito dentro dos parênteses de uma função ou método).

Vamos fazer isso agora:

```python
input("Digite o seu nome: ")
```

Coloquei dentro dos parâmetros da função input() uma string que guia o usuário com o que ele deve fazer.

Mas, como não armazenei em nenhum lugar, o Python pegou a informação, mas logo em seguida a excluiu. Isso porque não esta sendo armazenada ou trabalhada em nenhum lugar.

Vamos fazer isso agora:

```python
nome = input("Digite o seu nome:")
```

Vamos ver se, dessa vez, o Python armazenou o valor que o usuário deu para a função input():

```python
print(nome)
```

Armazenou e a retornou corretamente.
