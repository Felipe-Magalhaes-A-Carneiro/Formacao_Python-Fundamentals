## ***12. Muitos valores para muitas variáveis***

É possível atribuir valores para multiplas variaveis em uma só linha:

```python
a, b, c = "a de abelha", "b de baleia", "c de casa"

print(a)
print(b)
print(c)
```

----------------------------------------------------------------------------------------------------

## ***13. Um valor para muitas variáveis:***

Você pode também atribuir o mesmo valor para multiplas variáveis em uma só linha:

```python
a = b = c = "todas são ABC"

print(a)
print(b)
print(c)
```

## ***Unpack a Collection***

Se você tiver uma lista, tupla em Python, é possível extrair seus valores/elementos para dentro de variaveis:

```python
frutas = ["maça em uma lista", "banana em uma lista", "laranja em uma lista"]

maca, banana, laranja = frutas

print(maca)
print(banana)
print(laranja)
```
