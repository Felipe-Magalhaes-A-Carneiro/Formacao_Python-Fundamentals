# ***funcoes e metodos***

----------------------------------------------------------------------------------------------------

As linguagens de programação utilizam funcões e métodos que nada mais são que blocos de códigos feitos para executar um conjunto de ações muito comuns ou uteis. A ideia é portanto a de facilitar a criaçaõ de códigos. Poupando muito, mas muito tempo de desenvolvimento. 

Para chamar esses blocos de códigos prontos tudo o que devemos fazer é chama-las pelo seu nome.


A forma de chama-la é:

*nome da função * + '( )'

Os desenvolvedores do Python sempre procuraram utilizar criar o nome das funções com os termos mais adequados para o que a função de fato faz, e portanto, o mais próximo possível da língua inglesa.

----------------------------------------------------------------------------------------------------

### ***print( )***

```python
print("Python")
```

Perceba que o que colocarmos dentro de seus parênteses (chamamos o que há dentro de 'argumentos' ou 'parâmetros') é o que o Python irá trabalhar para executar, segundo o nome utilizado.

Neste caso 'print', ou seja, é como se falassemos: "imprima(isto)". O Python executará perfeitamente.

## ***Métodos***

----------------------------------------------------------------------------------------------------

Segundo o professor Anderson, do SENAI: "Métodos nada mais são que funções com donos".

Reparei que ele tinha toda a razão. 

Essa é a melhor forma de descrever o que são métodos, de fato.

Além disso:

Métodos são, inclusive, atributos que estão ligados diretamente ao tipo de dado que queremos utilizar com ele e que queremos que algo aconteça com ele.

Ou seja: 

- Se estamos querendo alterar uma string, logo, temos que utilizar os métodos que uma string é capaz / permite de ser utilizada;

- O mesmo para o tipo de dado int, alguns métodos dele não funcionam em strings e vice-versa.

Tudo por conta do tipo de dado que ele é.

## ***Métodos para Strings***

```python
"Felipe" # utilizaremos esta string com o meu nome para muda-lo
```

```python
"Felipe".upper()
```

Repare que, como disse anteriormente, como utilizamos uma string apenas os métodos do tipo de dado string poderam ser aceitos. Os métodos do tipo de dado int, por exemplo, não serão aceitos pelo Python.

```python
nome  = "Felipe"
```

Criamos uma string e armazenamos ela na variavel 'nome'. Temos o mesmo resultado que o anterior.

```python
nome.upper()
```

## ***Métodos para Inteiros***

### ***.bit_length( )***

Retorna o número de bits necessários para representar um inteiro em binário, excluindo o sinal e os zeros à esquerda.

```python
(10).bit_length()
```

Retorna 4 (pois 10 é 1010 em binário)
