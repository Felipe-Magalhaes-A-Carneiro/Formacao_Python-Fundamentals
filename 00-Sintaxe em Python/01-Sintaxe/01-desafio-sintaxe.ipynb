# 01 - Desafio
Crie um programa que:
- Pede pelo seu nome e idade;
- Dá 'Oi' para você;
- Conta quantas letras seu nome possui;
- Fala quantos anos você terá daqui a 5 anos.

```python
nome = input("Digite o seu nome: ")
idade = int(input("Digite a sua idade: "))
letras_em_nome = len(nome)
idade_mais_cinco = idade + 5

print(f"Oi, {nome}!")
print(f"Ei, {nome}, você sabia que o seu nome possui {letras_em_nome} letras?")
print(f"E que daqui a 5 anos você terá {idade_mais_cinco} anos de idade? Legal, né")
```

# 02 - DESAFIO RELÂMPAGO

- Faça um programa que pergunte ao usuário o nome do produto, preço e o desconto.

- O programa, então, irá imprimir de forma clara e organizada as informações (Incluindo o preço final com desconto aplicado). 

- No início, deverá parecer o nome da loja e no final uma saudação, depois deverá uma mensagem de volte sempre.

```python
usuario_nome = input("Digite o seu nome: ")

produto_nome = input("Digite o nome do produto: ")
produto_preco = float(input("Digite o valor do produto em R$: ").title())
desconto = float(input("Digite o desconto que será aplicado no produto em %: "))

novo_preco = produto_preco - desconto

print(f"""  
********** BEM-VINDO(A) À LOJA ECLESIAL! **********

Detalhes do produto cadastrado:
Nome do produto: {produto_nome};
Preço: {produto_preco:.2f};
Desconto que será aplicado: {desconto}%;

   O PRODUTO "{produto_nome.upper()}" ESTÁ COM O SEU PREÇO ATUALIZADO!    

O produto {produto_nome} agora está com uma promoção de {desconto}% APLICADO. 

Valor do produto para compra é de R${novo_preco}!

********** OBRIGADO PELA VISITA. VOLTE SEMPRE! ********** 

""")
```
