# ***Variaveis:***

----------------------------------------------------------------------------------------------------

Variaveis precisam de um nome e, em seguida, do sinal de 'atribuição' ( = ) e, então, o valor/elemento/conteúdo que você quer.

Imagine que você precise escrever um grande texto (string) e não só terá que faze-lo naquele momento, como também, traze-lo novamente em diferentes partes do seu código. Você pode fazer isso de diferentes modos:

1. Escrever o texto todo novamente. Você pode fazer isso, mas será uma grande perda de tempo.
2. Copiar o texto e replica-lo. Você pode fazer isso, mas também perderá tempo e terá que procura-lo no seu código toda vez que quiser ele. Imagine que o seu código tenha 1.000 linhas de código. Não parece agradável, não é?

3. Ou você pode fazer a melhor coisa possível: Armazenar esse texto (ou qualquer outra coisa praticamente) em uma variável!

Veja o seguinte texto, que está entre três aspas duplas (utilizamos as três para indicar ao Python que queremos uma String que pulará linhas dentro dela e que ele deve imprimi-la exatamente assim). Vamos armazena-la em uma variavel:

```python
""" Quando a Indesejada das gentes chegar

(Não sei se dura ou caroável),

talvez eu tenha medo.

Talvez sorria, ou diga:

— Alô, iniludível!

O meu dia foi bom, pode a noite descer.

(A noite com os seus sortilégios.)

Encontrará lavrado o campo, a casa limpa,

A mesa posta,

Com cada coisa em seu lugar.

[*Opus 10*] """
```

```python
poema_consoada = """ Quando a Indesejada das gentes chegar

(Não sei se dura ou caroável),

talvez eu tenha medo.

Talvez sorria, ou diga:

— Alô, iniludível!

O meu dia foi bom, pode a noite descer.

(A noite com os seus sortilégios.)

Encontrará lavrado o campo, a casa limpa,

A mesa posta,

Com cada coisa em seu lugar.

[*Opus 10*] """
```

Vamos imprimir o poema "Consoada" para o usuário:

```python
print(poema_consoada)
```

Ou seja, sempre que quisermos usar o poema "Consoada" basta utilizarmos a variável que criamos (utilizando o nome que demos para ela) e então todo o conteúdo que foi armazenado nela será retornado pelo Python!

# ***Nomenclatura e regras***

----------------------------------------------------------------------------------------------------

No universo Python, a nomenclatura de variáveis não é apenas uma questão de preferência estética, mas segue um conjunto rigoroso de regras sintáticas e convenções de estilo que visam a legibilidade e a manutenção do código.

## ***1. Regras Obrigatórias de Nomenclatura***

Para que um código seja considerado válido pelo interpretador, os nomes das variáveis (identificadores) devem obedecer a três regras fundamentais:

- Composição de caracteres: Devem conter apenas letras (maiúsculas ou minúsculas), números e o caractere de sublinhado (underscore _).

- Início do nome: Devem obrigatoriamente começar com uma letra ou um underscore, ***nunca*** com um número.

- Palavras Reservadas: É proibido utilizar palavras que o Python reserva para sua própria estrutura funcional, como `if`, `else`, `while`, `def`, `class`, entre outras.

### ***1.1 case-sensitive***

Além disso, a linguagem é case-sensitive, o que significa que `spam`, `SPAM`, `Spam` e `sPaM` são tratadas como quatro variáveis distintas e independentes

----------------------------------------------------------------------------------------------------

## ***2. Convenções e Estilo (PEP 8)***

Para além do que é "permitido", a comunidade Python segue o guia de estilo PEP 8, que padroniza a escrita para humanos.

- Variáveis e Funções: Devem usar o padrão `lower_case_with_underscores` (letras minúsculas separadas por sublinhado).

- Classes: Devem seguir o padrão `CamelCase` ou `CamelCaps` (cada palavra começando com maiúscula, sem sublinhados).

- Nomes Mnemônicos: As fontes recomendam escolher nomes que documentem o uso da variável (ex: `salario` é melhor que `s`), ajudando a lembrar o propósito daquele dado.

- Evitar confusão visual: Deve-se ter cautela ao usar a letra minúscula `l` e a maiúscula `O`, pois podem ser facilmente confundidas com os números `1` e `0`.

----------------------------------------------------------------------------------------------------

## ***3. O Contexto de Variáveis e Dados***

Conceitualmente, uma variável em Python funciona como uma etiqueta ou caixa que aponta para um valor armazenado na memória do computador.

- Tipagem Dinâmica e Forte: Python não exige que você declare o tipo da variável (como `int` ou `str`) antes de usá-la; o tipo é definido automaticamente no momento da atribuição (usando o operador `=`).

- Mutabilidade: É crucial entender que alguns tipos de dados são imutáveis (como strings e tuplas), o que significa que seus valores não podem ser alterados após a criação — apenas substituídos por um novo objeto. Já as listas são mutáveis, permitindo adicionar ou remover itens diretamente no objeto original.

- Privacidade de Atributos: Por convenção, nomes que começam com um único underscore (ex: `self._name`) indicam que aquele dado é de uso interno ("privado") da classe e não deve ser acessado diretamente por outros módulos.

----------------------------------------------------------------------------------------------------

## ***4. Revisão Edição 2026 (Documentação Oficial 3.14.5)***

Ao revisar estas regras sob a ótica da Documentação 3.14.5, observamos que a fundação da nomenclatura permanece estável, mas o contexto de uso dos dados evoluiu:

- Novas Palavras Reservadas: A lista de palavras-chave foi atualizada em relação aos livros mais antigos, incluindo termos como `async` e `await` para programação assíncrona, e o suporte ao Pattern Matching com a instrução `match`, que introduz novos contextos de nomes de variáveis em padrões.

- f-strings como Padrão: Enquanto fontes antigas discutem o operador `%` para formatar nomes em strings, a Documentação 3.14.5 formaliza e aprimora as f-strings (PEP 701) como a forma moderna e preferencial de interpolar dados e nomes de variáveis em textos.

- Anotações de Tipo: Em 2026, o uso de anotações de tipo (ex: `nome: str = "José"`) tornou-se uma prática recomendada para grandes projetos, auxiliando depuradores e ferramentas de análise estática, algo que é apenas mencionado brevemente em edições anteriores.
