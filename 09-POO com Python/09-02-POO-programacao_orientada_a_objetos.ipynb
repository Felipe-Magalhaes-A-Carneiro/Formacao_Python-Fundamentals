# ***A Programação Orientada a Objetos (POO)*** 

É um paradigma de programação que organiza o código em "objetos" que combinam dados (características) e comportamentos (ações).

Imagine que você está criando um jogo. Em vez de ter variáveis soltas para vida_heroi, forca_heroi e funções como atacar(personagem), você cria um molde chamado Guerreiro que já contém tudo isso.

----------------------------------------------------------------------------------------------------

## ***1. Os Quatro Pilares da POO***
Para dominar a POO, você precisa entender estes quatro conceitos fundamentais:

### A. Abstração
É simplificar um objeto real para o que importa no seu sistema.

- Exemplo: Para um sistema bancário, um "Cliente" precisa de nome e CPF. Para um hospital, ele precisa de tipo sanguíneo. A abstração filtra o que é irrelevante.

### B. Encapsulamento
É "esconder" os detalhes internos de como um objeto funciona e proteger seus dados. Em Python, usamos um prefixo de sublinhado (_ ou __) para indicar que algo é privado.

- Exemplo: Você não mexe direto nas engrenagens de um relógio; você usa os botões externos (métodos) para ajustar a hora.

### C. Herança
Permite que uma classe (Filha) herde atributos e métodos de outra (Pai), evitando repetição de código.

- Exemplo: Uma classe Carro e uma classe Moto podem herdar de uma classe Veiculo.

### D. Polimorfismo
É a capacidade de diferentes classes responderem à mesma mensagem de formas diferentes.

- Exemplo: Tanto Cachorro quanto Gato têm o método falar(), mas um faz "Au" e o outro "Miau".

----------------------------------------------------------------------------------------------------

## ***2. A Anatomia de uma Classe em Python***
Vamos construir uma classe que aplique esses conceitos:

```python
class Veiculo: # Classe Pai
    # caracteristicas:
    def __init__(self, marca, modelo):
        self.marca = marca
        self.modelo = modelo
        self._ligado = False # Encapsulamento (privado por convenção)

    # atributos:
    def ligar(self):
        self._ligado = True
        print(f"{self.modelo} ligou.")

class Carro(Veiculo): # Herança: Carro herda de Veiculo
    def mover(self): # Método específico
        if self._ligado:
            print(f"O carro {self.modelo} está andando.")
        else:
            print("Ligue o carro primeiro!")

class Aviao(Veiculo):
    def mover(self): # Polimorfismo: mover() aqui é voar
        print(f"O avião {self.modelo} está decolando!")
```

# ***3. O Método Mágico _ _str_ _***

Lembra que falamos que tudo em Python é praticamente um objeto? Quando você dá um print(objeto), o Python procura o método __str__ para saber como exibir aquele objeto em texto.

```python
class Livro:
    def __init__(self, titulo, autor):
        self.titulo = titulo
        self.autor = autor

    def __str__(self):
        return f"'{self.titulo}' por {self.autor}"

meu_livro = Livro("Dom Casmurro", "Machado de Assis")
print(meu_livro) # Saída: 'Dom Casmurro' por Machado de Assis
```

# ***O que é 'self'?***

É uma referencia ao objeto, chamamos de 'referência explícita'.

O self é, sem dúvida, o conceito que mais confunde quem está começando em POO, mas a explicação é mais simples do que parece: ***o self é o nome que o objeto usa para se referir a si mesmo***.

Imagine que você tem uma fábrica de carros (a Classe). Quando um carro sai da linha de montagem (o Objeto), ele precisa de uma forma de saber qual é a sua própria cor, sem confundir com a cor do carro que saiu antes dele.

# ***1. Por que o self é necessário?***

O Python não "mágica" os atributos para dentro do objeto. Quando você define um método, o Python envia automaticamente o próprio objeto como o primeiro argumento. O self é o "balde" que recebe esse objeto.

Veja a diferença:

```python
class Celular:
    def __init__(self, marca, modelo):
        # "Neste objeto específico (self), guarde esta marca"
        self.marca = marca 
        self.modelo = modelo

    def ligar(self):
        # Sem o self, o Python não saberia DE QUEM é esse modelo
        print(f"O {self.modelo} está ligando...")

# Criando instâncias (objetos)
meu_celular = Celular("Apple", "iPhone 15")
seu_celular = Celular("Samsung", "S24")

meu_celular.ligar() # Saída: O iPhone 15 está ligando...
seu_celular.ligar() # Saída: O S24 está ligando...
```

# ***2. O que acontece "por baixo dos panos"?***

Quando você chama meu_celular.ligar(), o Python traduz isso internamente para:
Celular.ligar(meu_celular)

O objeto meu_celular entra no lugar do parâmetro self. É por isso que, embora você defina o método com self, você não passa nada para ele na hora de chamar a função.

# ***3. O self é obrigatório?***

- O nome self: Não é uma palavra-chave protegida. Você poderia usar banana ou eu, e o código funcionaria. Mas ***99,9% dos programadores Python usam*** self. Usar outro nome é como dirigir do lado errado da pista: você pode até conseguir, mas vai causar um acidente na leitura do código.

- ***A presença do parâmetro***: Sim, é obrigatória para métodos de instância. Se você esquecer o self na definição:
def ligar(): print("Oi")

E tentar chamar meu_celular.ligar(), o Python vai dar um erro dizendo que você passou 1 argumento, mas a função aceita 0.

# ***4. Analogia do Crachá***

Pense no 'self' como um crachá.

- A classe é o manual de instruções da empresa.

- O objeto é o funcionário.

- O self é o funcionário olhando para o próprio crachá para dizer: "Eu sou o funcionário número 42 e meu departamento é Vendas".
