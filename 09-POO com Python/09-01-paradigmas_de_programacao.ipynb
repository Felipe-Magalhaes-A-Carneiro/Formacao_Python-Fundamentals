# ***O que são 'Paradigmas de Programação?***

Entender os paradigmas de programação é como entender os diferentes estilos de culinária: você pode fazer um ovo usando técnicas francesas, italianas ou japonesas. O resultado final é um ovo, mas o processo de pensamento e as ferramentas são diferentes.

Um paradigma é, essencialmente, uma abordagem ou estilo de resolver problemas através do código. O Python é uma linguagem multiparadigma, o que significa que ele permite que você misture esses estilos conforme a necessidade.

# 1. Paradigma Imperativo - (Código linha por linha)

É o "estilo receita de bolo". Você foca em como o computador deve fazer algo, dando uma sequência de comandos que alteram o estado do programa passo a passo.

- Foco: Instruções diretas e mudança de variáveis.

- Analogia: "Pegue a chave, coloque na ignição, gire, pise no acelerador."

# 2. Paradigma Procedural - (Código com uso de funções)

É uma evolução do imperativo. Em vez de um bloco gigante de código, você organiza as instruções em procedimentos (que no Python chamamos de funções).

- Foco: Divisão do problema em funções reutilizáveis.

- Vantagem: Facilita a manutenção. Se o cálculo do imposto mudar, você só altera a função calcular_imposto() e não o código inteiro.

# 3. Paradigma Funcional

Aqui, o programa é visto como uma série de transformações matemáticas. Em vez de mudar o valor de uma variável (estado), você passa um dado por uma função e recebe um novo dado.

- Imutabilidade: Você evita mudar dados existentes. Você cria novos.

- Funções de Primeira Classe: Como vimos, você passa funções para outras funções (como o map, filter e reduce).

- Analogia: Uma linha de montagem onde o produto entra de um jeito e sai transformado, mas a máquina em si não muda.

```python
# Estilo Funcional em Python
numeros = [1, 2, 3]
dobro = list(map(lambda x: x * 2, numeros)) # Não mudamos 'numeros', criamos 'dobro'
```

# 4. Paradigma Orientado a Eventos

Neste estilo, o fluxo do programa é determinado por eventos: cliques de mouse, mensagens de outros programas, ou sensores. O programa fica "ouvindo" e reage quando algo acontece.

- Loop de Eventos: O programa roda um ciclo infinito esperando uma ação.

- Callback: É a função que você define para ser executada quando o evento ocorrer.

- Onde é usado: Interfaces gráficas (botões), servidores web e jogos.

Imagine um garçom: ele não fica te perguntando "quer comer?" a cada segundo (isso seria um loop imperativo). Ele fica parado e reage quando você levanta a mão (isso é Orientado a Eventos).
