# ***Polimorfismo***

----------------------------------------------------------------------------------------------------

A palavra vem do grego e significa "muitas formas".

No Python, o polimorfismo permite que objetos de diferentes classes sejam tratados da mesma maneira se eles compartilharem a mesma interface (ou seja, se tiverem métodos com o mesmo nome).

----------------------------------------------------------------------------------------------------

## ***1. O Conceito Principal***

Imagine que você tem um controle remoto universal. Você aperta o botão "Play" e:

- Se for um DVD, ele gira o disco.

- Se for o Netflix, ele inicia o streaming.

- Se for o Spotify, ele toca a música.

O botão é o mesmo (a interface), mas a ação final depende de qual objeto está sendo controlado (o comportamento).

----------------------------------------------------------------------------------------------------

## ***2. Polimorfismo com Herança***

É o uso mais comum. Uma classe pai define um método, e as classes filhas o "reescrevem" (sobrescrita) para agir de forma específica.

```python
class Animal:
    def fazer_som(self):
        pass

class Cachorro(Animal):
    def fazer_som(self):
        return "Au Au!"

class Gato(Animal):
    def fazer_som(self):
        return "Miau!"

# Uma função que não se importa com o tipo do animal, apenas que ele faça som
def emitir_som(animal):
    print(animal.fazer_som())

animais = [Cachorro(), Gato()]

for a in animais:
    emitir_som(a)
```

Aqui, a função emitir_som é polimórfica: ela funciona com qualquer objeto que herde de Animal.

----------------------------------------------------------------------------------------------------

## ***3. Duck Typing: O Polimorfismo "Estilo Python"***

O Python leva o polimorfismo além da herança com o conceito de Duck Typing ("Tipagem de Pato"). A regra é:

**"Se anda como um pato e grasna como um pato, então é um pato."**

Em Python, você não precisa herdar de uma classe específica para ser polimórfico. Se o seu objeto tem o método que eu preciso, eu o uso.

```python
class Pato:
    def quack(self):
        print("Quack!")

class Pessoa:
    def quack(self):
        print("Eu estou imitando um pato!")

def fazer_quack(objeto):
    objeto.quack()

fazer_quack(Pato())   # Funciona
fazer_quack(Pessoa()) # Também funciona! Mesmo que Pessoa não seja um Animal.
```

----------------------------------------------------------------------------------------------------

## ***4. Polimorfismo em Funções Integradas***

Você já usa polimorfismo no Python sem saber! A função len(), por exemplo, é polimórfica:

- len("Python") -> Conta caracteres em uma String.

- len([1, 2, 3]) -> Conta itens em uma Lista.

- len({"a": 1}) -> Conta chaves em um Dicionário.

O comportamento da função muda conforme o tipo de objeto que você entrega a ela.

----------------------------------------------------------------------------------------------------

## ***5. Por que o Polimorfismo é útil?***

1. Flexibilidade: Você pode criar novas classes (ex: um Robo que faz som) e as funções antigas continuarão funcionando sem precisar de alteração.

2. Código Limpo: Você evita usar infinitos if e else para verificar o tipo do objeto (if tipo == 'Cachorro'...). Você simplesmente chama o método e deixa o objeto se virar.

----------------------------------------------------------------------------------------------------

## ***Resumo:***

Herança: Define o que um objeto é.

Polimorfismo: Define o que um objeto faz de forma diferente, mesmo recebendo a mesma ordem.
