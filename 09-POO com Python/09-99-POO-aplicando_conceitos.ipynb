# Primeiro programa POO

João, tem uma bicicletaria e gostaria de registrar as vendas de suas bicicletas. Crie um programa onde João inform: cor, modelo, ano e valor da bicicleta vendida. Uma bicicleta pode: buzinar, parar e correr. Adicione esses comportamentos!

obs.: o professor não utilizou o conceito de inicializador, por enquanto.

```python
class Bicicleta: 
    # caracteristicas:
    def __init__(self, cor, modelo, ano, valor):
        self.cor = cor
        self.modelo = modelo
        self.ano = ano
        self.valor = valor
        
    # atributos/ métodos:
    def buzinar(self):
        print("Plim... plim...")

    def parar(self):
        print("Freiando a bicicleta...")
        print("A bicicleta parou.")

    def correr(self):
        print(f"A bicicleta {self.modelo} esta acelerando cada vez mais...")
```

# Instanciando a classe e criando o objeto:

```python
bicicleta_1 = Bicicleta("vermelha", "Caloi", 2022, 1200)
```

utilizando o objeto e seus atributos:

```python
bicicleta_1.correr()
```

```python
bicicleta_1.buzinar()
```

```python
bicicleta_1.parar()
```

Criando outras bicicletas utilizando o molde da classe Bicicleta:

- Instanciando o objeto em uma variável:

```python
bicicleta_2 = Bicicleta("azul", "Sense", 2022, 1200)
```

```python
bicicleta_2.correr()
```

# Outra perspectiva para aprendermos de vez...

Podemos chamar diretamente o Objeto e depois um atributo dele (para lembrarmos, isso foi criado na Classe Bicicleta):

```python
Bicicleta.buzinar()
```

Perceba que ele aceita, mas diz que há um erro.

Ele diz que falta um argumento posicional, que neste caso é o 'self', que aqui é uma das instancias que fizemos antes. Faça da seguinte maneira e estará correto:

```python
Bicicleta.buzinar(bicicleta_1)
```

ou seja, o que fizemos acima é o equivalente de fazermos dessa forma:

```python
bicicleta_1.buzinar()
```

Aplique o seguinte diagrama em uma contrução de classes real em Python:

```python
# Criando Clase Pai (Superclasse)

class Veiculo:
    def __init__(self, tipo, cor, placa, n_rodas):
        self.tipo = tipo
        self.cor = cor
        self.placa = placa
        self.n_rodas = n_rodas


    # atributos
    def ligar_motor(self):
        print(f"{self.tipo} esta ligando")

    def acelerar(self):
        print(f"{self.tipo} esta acelerando...")

    def bozinar(self):
        print(f"{self.tipo} esta bozinando!")

    def freiar(self):
        print(f"{self.tipo} freiou!")
    
    def estacionar(self):
        print(f"{self.tipo} estacionaou.")
    

# (Subclasse)
class Motocicleta(Veiculo):

    pass

# (Subclasse)
class Carro(Veiculo):

    pass

# (Subclasse)
class Caminhão(Veiculo):
    def __init__(self, tipo, cor, placa, n_rodas, carregado):
    
        super().__init__(tipo, cor, placa, n_rodas)
        self.carregado = carregado


    def esta_carregado(self):
        print(f"O {self.tipo} {'esta sim' if self.carregado else 'não esta'} carregado")
```

```python
moto = Motocicleta("moto", "preta", "AAB-152", 2) 
```

A Classe Motocicleta herdou todas as caracteristicas e atributos da classe Veículo. 

```python
moto.ligar_motor()
```

```python
carro = Carro("carro", "vermelho", "BCA-214", 4)
```

```python
carro.ligar_motor()
```

```python
caminhao = Caminhão("caminhão", "branco", "CFB-456", 8, True)
```

```python
caminhao.esta_carregado()
```

Criamos o método 'esta_carregado()' para o caminhão. 

Mas será que podemos fazer com que o objeto 'carro', por exemplo, possa também ter esse método, já que estamos utilizando a herança multipla?

Vamos ver:

```python
carro.esta_carregado()
```

Perceba que não. 

Não é possível o objeto 'carro' que tem a sua origem na classe 'Carro', este último que possuí uma herença da Classe SuperClasse 'Veiculos', ter acesso ao método 'esta_carregado()' da classe 'Caminhão', mesmo que estes tenham essa condição de receberem a herença da Super Classe 'Veiculos'.

## Colocando um inicializador na sub classe 'Caminhão'.

### Vamos aprender o que é 'Sobreescrever uma implementação de um método':

Vamos fazer com que o método 'esta_carregado()' receba uma alteração em sua condição. Para isso, precisamos colcoar o inicializador _ _init_ _() da seguinte maneira:

Obs.: Para fins didáticos, vamos trazer a seguir de forma comentada a parte específica do código que retrata o que essa parte do código esta fazendo.

Versão antes do inicializador:

```python
# (Subclasse)
# class Caminhão(Veiculo):
#     def esta_carregado(self):
#         print(f"O {self.tipo} não esta carregado")
```

Versão com o inicializador:

```python
# class Caminhão(Veiculo):
#     def __init__(self, carregado):
#         self.carregado = carregado


#     def esta_carregado(self):
#         print(f"O {self.tipo} {'sim esta' if self.carregado else 'não esta'} carregado")
```

Se eu tenhar utilizar o meu objeto 'caminhao' como ele esta, o Python irá dar um erro a respeito de argumentos.

Ele dirá que eu estou tentando usa-lo sem dar nenhum argumento. Dizendo que a subclasse 'Caminhão' exige 5 argumentos.

Como assim 5 argumentos? Bom, 4 argumentos são da super classe "Veiculos" e 1 é do argumento que acabamos de criar ao adicionar: "carregado", este que esta naquele mais um construtor/inicializador _ _init_ _() dentro da subclasse 'Caminhao'.

***Como resolver isso?***

Uma das formas de resolver, é simplismente trazendo os argumentos do primeiro _ _init_ _() da super classe 'Veiculo' para a subclasse 'Caminhão':


```python
# # (Subclasse)
# class Caminhão(Veiculo):
#     def __init__(self, tipo, cor, placa, n_rodas, carregado):
#         self.tipo = tipo
#         self.cor = cor
#         self.placa = placa
#         self.n_rodas = n_rodas
#         self.carregado = carregado


#     def esta_carregado(self):
#         print(f"O {self.tipo} {'sim esta' if self.carregado else 'não esta'} carregado")
```

Feita as devidas alterações, devemos criar o objeto e agora com o atributo se 'sim', ou 'não'

```python
# caminhao = Caminhão("caminhão", "branco", "CFB-456", 8, "sim")
```

```python
caminhao.esta_carregado()
```

Essa ação é chamada de 'Sobreescrever uma implementação de um método'. O Python permite que isso aconteça

Mas existe outra forma de fazermos isso e é com o uso da palabra reservada Super:

## Super( ) - Vamos implemantar o Super para podermos trazer os atributos da Super classe para a sua sub classe filha:
