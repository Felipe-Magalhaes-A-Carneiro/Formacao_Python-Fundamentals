# Primeiro programa POO

João, tem uma bicicletaria e gostaria de registrar as vendas de suas bicicletas. Crie um programa onde João inform: cor, modelo, ano e valor da bicicleta vendida. Uma bicicleta pode: buzinar, parar e correr. Adicione esses comportamentos!

obs.: o professor não utilizou o conceito de inicializador, por enquanto.

```python
class Bicicleta: 
    # caracteristicas:
    def __init__(self, cor, modelo, ano, valor):
        self.cor = cor
        self.modelo = modelo
        self.ano = ano
        self.valor = valor
        
    # atributos/ métodos:
    def buzinar(self):
        print("Plim... plim...")

    def parar(self):
        print("Freiando a bicicleta...")
        print("A bicicleta parou.")

    def correr(self):
        print(f"A bicicleta {self.modelo} esta acelerando cada vez mais...")
```

# Instanciando a classe e criando o objeto:

```python
bicicleta_1 = Bicicleta("vermelha", "Caloi", 2022, 1200)
```

utilizando o objeto e seus atributos:

----------------------------------------------------------------------------------------------------

```python
bicicleta_1.correr()
```

```python
bicicleta_1.buzinar()
```

```python
bicicleta_1.parar()
```
