# ***@property***

----------------------------------------------------------------------------------------------------

O @property é um dos decoradores mais "mágicos" do Python. Ele é a ferramenta oficial para implementar o ***Encapsulamento*** de forma elegante.

Basicamente, ele transforma um método (uma função da classe) em um atributo (uma variável da classe). Isso permite que você execute uma lógica (como uma validação) no momento em que alguém tenta ler ou alterar um valor, mas sem que o usuário da classe precise usar parênteses ().

## ***1. O Problema: Atributos expostos***

Imagine que você tem uma classe Conta. Se o saldo for um atributo comum, qualquer um pode colocar um valor negativo nele:

```python
class Conta:
    def __init__(self, saldo):
        self.saldo = saldo

c = Conta(100)
c.saldo = -5000  # O sistema aceita, mas isso é um erro de negócio!

print(c.saldo)
```

----------------------------------------------------------------------------------------------------

## ***2. A Solução: O Decorador @property***

Com o @property, você esconde o atributo real (usando _ ou __) e cria métodos para acessá-lo.

### ***O "Getter" (Leitura)***

O primeiro passo é criar o método que retorna o valor. Ele deve ter o nome que você quer que o atributo tenha.

```python
class Conta:
    def __init__(self, saldo):
        self._saldo = saldo # Atributo "privado"

    @property
    def saldo(self):
        print("Acessando o saldo...")
        return self._saldo
```

Agora, você acessa como print(c.saldo), e não como c.saldo().

----------------------------------------------------------------------------------------------------

## ***3. O "Setter" (Escrita)***

Para permitir que o valor seja alterado com uma regra de segurança, usamos o .setter.

```python
@saldo.setter
    def saldo(self, novo_valor):
        if novo_valor < 0:
            print("Erro! O saldo não pode ser negativo.")
        else:
            self._saldo = novo_valor

# Uso:
c = Conta(100)
c.saldo = 500   # O setter é chamado automaticamente
c.saldo = -50   # O setter bloqueia a operação!
```

----------------------------------------------------------------------------------------------------

## ***4. Vantagens Reais***

- ***Compatibilidade Retroativa:*** Se você começou o projeto com um atributo simples (self.saldo) e depois percebeu que precisava de validação, você pode transformar em @property sem quebrar o código de quem já usa sua classe.

- ***Atributos Somente Leitura:*** Se você criar apenas o @property e não criar o .setter, o atributo se torna impossível de alterar por fora.

- ***Atributos Calculados:*** Você pode ter uma propriedade area que não existe no banco de dados, mas é calculada na hora a partir de largura * altura.

----------------------------------------------------------------------------------------------------

## ***Resumo da Sintaxe***

1. @property: Define o método que será o "Getter" (pegar valor).

2. @nome_do_metodo.setter: Define o método que será o "Setter" (definir valor).

3. @nome_do_metodo.deleter: (Menos comum) Define o que acontece ao usar del objeto.nome.

----------------------------------------------------------------------------------------------------


## ***Por que não usar get_saldo() e set_saldo() como no Java?***
Em Python, isso é considerado feio (não-pythonico). O Python prefere que o acesso aos dados pareça natural (objeto.atributo), e o @property permite que esse acesso natural seja inteligente e seguro por baixo dos panos.

# ***O forte do @property = atributos calculados***

Um dos usos mais poderosos do @property é criar atributos calculados. Eles não ocupam espaço na memória do objeto (não são guardados em uma variável), mas são gerados "on-the-fly" (na hora) sempre que você os chama.

Imagine que você tem um personagem de RPG com força e agilidade. O poder de ataque dele depende desses dois valores. Se você guardar o "poder" em uma variável comum, terá que se lembrar de atualizá-la toda vez que a força aumentar. Com o @property, isso é automático.

----------------------------------------------------------------------------------------------------

```python
class Personagem:
    def __init__(self, nome, forca, agilidade):
        self.nome = nome
        self.forca = forca
        self.agilidade = agilidade

    @property
    def poder_ataque(self):
        # O cálculo acontece toda vez que você der um print(p.poder_ataque)
        print(f"--- Calculando poder de {self.nome} ---")
        return (self.forca * 2) + self.agilidade

# Criando o herói
heroi = Personagem("Arthur", 10, 5)

# Acessando o atributo calculado
print(f"Poder atual: {heroi.poder_ataque}") # Saída: 25

# Se a força aumentar, o poder muda sozinho!
heroi.forca = 20
print(f"Novo poder: {heroi.poder_ataque}") # Saída: 45
```

----------------------------------------------------------------------------------------------------

## ***Por que isso mostra a potência do @property?***

1. Sincronização Perfeita: Não existe o risco de o poder_ataque estar "desatualizado" em relação à forca.

2. Interface Limpa: Para quem usa sua classe, poder_ataque parece uma variável comum. Eles não precisam saber que existe uma fórmula matemática rodando por trás.

3. Segurança: Como não criamos um .setter para o poder_ataque, ninguém pode fazer heroi.poder_ataque = 9999. O valor é protegido pela lógica do código.

----------------------------------------------------------------------------------------------------

## ***Quando usar Atributos Calculados?***

- Geometria: Calcular a area ou o perimetro a partir de largura e altura.

- Finanças: Calcular o valor_com_imposto a partir do preco_base.

- Data/Hora: Calcular a idade a partir da data_nascimento.

- Jogos: Calcular o status_critico baseado na sorte do jogador.
