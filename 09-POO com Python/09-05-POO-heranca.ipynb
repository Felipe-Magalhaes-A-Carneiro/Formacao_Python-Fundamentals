# ***Herança***

----------------------------------------------------------------------------------------------------

Quando você usa o termo:

"A Classe esta se estendendopara * outra classe *."

Quer dizer que a classe esta herdando (se estendendo)

```python
class A:

    pass

class B(A): # classe B é filha da classe A

    pass
```

## Benefícios

- Representa bem os relacionamentos do mundo real.

- Fortalece reutilização de código, não precisamos escrever o mesmo código repetidamente. Além disso, permite adicionar mais recuros a uma classe sem modificá-la.

- É de 'natureza transitiva', o que significa que, se a Classe B herdar da Classe A, todas as subclasses de B herdarão automaticamente da Classe A

# ***Herança Simples***

----------------------------------------------------------------------------------------------------

A Herança Simples é um dos pilares da Programação Orientada a Objetos (POO) e permite que uma classe (chamada de Filha ou Subclasse) herde todos os atributos e métodos de outra classe (chamada de Pai, Mãe ou Superclasse).

## A ideia Principal

A ideia principal é o reuso de código e a criação de uma hierarquia lógica. Se algo "é um" tipo de outra coisa (ex: um Gato é um Animal), ele pode herdar suas características.

### 1. Como funciona na prática?
Para herdar, basta colocar o nome da classe pai entre parênteses na definição da classe filha.

```python
# Classe Pai (Superclasse)
class Animal:
    def __init__(self, nome):
        self.nome = nome

    def comer(self):
        print(f"{self.nome} está comendo...")

# Classe Filha (Subclasse) - Herança Simples
class Cachorro(Animal):
    def latir(self):
        print(f"{self.nome} diz: Au Au!")

# Usando as classes
rex = Cachorro("Rex")
rex.comer() # Método herdado de Animal
rex.latir() # Método próprio de Cachorro
```

### 2. O papel do super()

Muitas vezes, a classe filha precisa adicionar algo novo ao "nascimento" (construtor), mas ainda quer manter o que o pai já fazia. Para isso, usamos a função super().

O super() serve para chamar métodos da classe pai dentro da classe filha.

```python
class Funcionario:
    def __init__(self, nome, salario):
        self.nome = nome
        self.salario = salario

class Gerente(Funcionario):
    def __init__(self, nome, salario, bonus):
        # Chama o construtor do pai para setar nome e salario
        super().__init__(nome, salario)
        self.bonus = bonus # Atributo exclusivo do Gerente

    def exibir_total(self):
        print(f"Total: R$ {self.salario + self.bonus}")

gerente_atual = Gerente("Ricardo", 6000.00, 700.00)

gerente_atual.exibir_total()
```

### 3. Sobrescrita de Método (Method Overriding)

Uma classe filha pode herdar um método, mas decidir que quer fazer as coisas de um jeito diferente. Basta definir um método com o mesmo nome na classe filha.

```python
class Passaro(Animal):
    def mover(self):
        print("O pássaro está voando.")

class Peixe(Animal):
    def mover(self):
        print("O peixe está nadando.")
```

### 4. Vantagens da Herança Simples

- Economia de tempo: Você não precisa reescrever o código que já existe na classe pai.

- Manutenção fácil: Se você descobrir um erro no método comer() de Animal, você conserta em um lugar só e todos os animais (Cachorro, Gato, Peixe) serão corrigidos automaticamente.

- Organização: Cria uma estrutura clara de categorias e subcategorias.

### 5. Resumo da terminologia

- Superclasse / Classe Base / Pai: A classe que doa as características.

- Subclasse / Classe Derivada / Filha: A classe que recebe as características.

- Herança Simples: Quando uma classe filha herda de apenas uma classe pai. (Existe a herança múltipla, mas ela é mais complexa e o Python também suporta!).

# ***Herança Multipla***

----------------------------------------------------------------------------------------------------

Obs.: Nem todas as línguagens possuem esse tipode de herança.

A ***Herança Múltipla*** acontece quando uma classe filha herda características de duas ou mais classes pai simultaneamente. No Python, isso é perfeitamente possível e muito potente, embora exija um cuidado redobrado para não virar uma "bagunça".

Imagine que você tem a classe Relogio e a classe Calendario. Se você criar uma classe Smartwatch, ela pode herdar de ambas, pois possui as funcionalidades de ambas as categorias.

### 1. Sintaxe da Herança Múltipla

Basta listar as classes pai separadas por vírgula dentro dos parênteses:

```python
class A:

    pass

class B: 

    pass

class C(A, B): # classe C é filha das classes A e B

    pass
```

```python
class Relogio:
    def mostrar_hora(self):
        print("Agora são 14:00")

class Calendario:
    def mostrar_data(self):
        print("Hoje é 06/03/2026")

# Herança Múltipla
class Smartwatch(Relogio, Calendario):
    pass

watch = Smartwatch()
watch.mostrar_hora() # Vem de Relogio
watch.mostrar_data() # Vem de Calendario
```
